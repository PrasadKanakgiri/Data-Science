# Insert/Delete in a Linked List

```python
class Node:
    
    def __init__(self, value):
        self.data = value
        self.next = None
    
class LinkedList():
    
    def __init__(self):
        self.head = None      # empty linked list
        self.n = 0            # total number of nodes

    # length of the linked list
    def __len__(self):
        return self.n
    
    # printing the linked list
    def __str__(self):
        curr = self.head
        
        result = ""
        while(curr != None):
            result = result + str(curr.data) + '->'
            curr = curr.next
            
        return result[:-2]  # -2 because we dont want to print '->' at the end of the linked list

    # insert at head
    def insert_head(self, value):
        # create a new node
        new_node = Node(value)
        
        # create connection
        new_node.next = self.head
        
        # reassign head
        self.head = new_node
        
        # increment n
        self.n = self.n + 1
    
    # insert at tail 
    def insert_tail(self, value):
        # create a new node
        new_node = Node(value)
        
        # if linked list is empty
        if self.head == Node:
            self.head = new_node
            self.n = self.n + 1
            return
        
        curr = self.head
        
        # travese till tail
        while(curr.next != None):
            curr = curr.next
        
        # create connection
        curr.next = new_node
        
        # increment n
        self.n = self.n + 1
        
    # insert at middle
    def insert_after(self, after, value):
        new_node = Node(value)
        
        curr = self.head
        
        while curr != None and curr.data != after:
            curr = curr.next
           
        # if num is not present in the list    
        if curr == None:
            return
        
        temp = curr.next
        curr.next = new_node
        new_node.next = temp
        
        # increment n
        self.n = self.n + 1
        
    # clear the linked list
    def clear(self):
        self.head = None
        self.n = 0
        
    # delete from head
    def delete_head(self):
        if self.head == None:
            return
        
        self.head = self.head.next
        
        self.n = self.n - 1
    
    # delete from tail
    def delete_tail(self):
        if self.head == None:
            return
        
        curr = self.head
        
        # only 1 element which is head
        if curr.next == None:
            return self.delete_head()
        
        while curr.next.next != None:
            curr = curr.next
        
        curr.next = None
        
        self.n = self.n - 1
    
    # delete from middle
    def delete_node(self, value):
        if self.head == None:
            return
        
        # If value is in head
        if self.head.data == value:
            self.delete_head()
            return
        
        curr = self.head
        
        # Find node before the node to delete
        while curr.next != None and curr.next.data != value:
            curr = curr.next
            
        # Value not found
        if curr.next == None:
            return
        
        temp = curr.next 
        curr.next = temp.next
        temp.next = None
        
        self.n = self.n - 1
    
    # def traverse(self):
    #     curr = self.head
        
    #     while(curr != None):
    #         print(curr.data)
    #         curr = curr.next
    
        
        
L = LinkedList()

L.insert_head(4)
L.insert_head(3)
L.insert_head(2)
L.insert_head(1)

# print(L)

L.insert_tail(5)
L.insert_tail(6)
# print(L)

L.insert_after(3, 30)
# print(L)
L.insert_after(5, 50)
# print(L)
L.insert_after(6, 60)
# print(L)

# L.clear()
print(L)
L.delete_head()
print(L)

L.delete_tail()
print(L)

L.delete_node(50)
print(L)
L.delete_node(30)
print(L)
L.delete_node(2)
print(L)


```
