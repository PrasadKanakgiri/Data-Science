### 43. Write  a program to print the following pattern 
```text
    * 
   *** 
  ***** 
 ******* 
*********
```

````python
# 1st solution
n = int(input("Enter number of rows: "))

for i in range(1, n + 1):
    print(" "*(n-i), end="")
    for j in range(1, i*2):
        print("*", end="")
    print()
    
    '''
stars                   spaces
    1  2 x 1 - 1        4   5 - 1
    3  2 x 2 - 1        3   5 - 2
    5  2 x 3 - 1        2   5 - 3
    7  2 x 4 - 1        1   5 - 4
    9  2 x 5 - 1        0   5 - 5
    
    '''
````

````python
# 2nd solution
n = int(input("Enter number of rows: "))

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
````
