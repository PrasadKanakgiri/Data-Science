### 12. Write a program to find the volume of the cylinder. Also find the cost when ,when the cost of 1litre milk is 40Rs

```python
# # V = πr2h
import math

r = float(input("Enter Radius of the circular base of the cylinder (in cm)"))
h = float(input("Enter height of the cylinder (in cm)"))

v = math.pi * r**2 * h

print(f"Volume of the cylinder with radius: {r}cm and height: {h}cm is {v:.1f}cm³")
```
