### 62. Write a python program to search a given number from a list 

```python
# 1st solution

l = [4, 2, 6, 8, 3 , 1, 10, 9]

search = 9

print(f"{search} is present at {l.index(search)}th index")

    
```

```python
# 2nd solution

l = [4, 2, 6, 8, 3 , 1, 10, 9]

search = 9

for i in range(len(l)):
    
    if l[i] == search:
        print(f"{search} is present at {i}th index")
        break

    
```

```python
# 3rd solution

l = [4, 2, 6, 8, 3 , 1, 10, 9]

search = 9

for index, value in enumerate(l):     # enumerate() gives you both the index and the value:
    
    if value == search:
        print(f"{search} is present at {index}th index")
        break
```
