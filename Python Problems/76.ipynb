### 76. Write a program to perform matrix multiplication on 2 matrices

```python
l1 = [
    [1, 2, 3],
    [4, 5, 6]
]

l2 = [
    [7, 8],
    [9, 10],
    [11, 12]
]



r1 = len(l1)
c1 = len(l1[0])

r2 = len(l2)
c2 = len(l2[0])

if c1 != r2:
    print("Matrix multiplication not possible")
else:
    l3 = []
    
    for i in range(r1):
        row = []
        
        for j in range(c2):
            total = 0
            
            for k in range(c1):
                total += l1[i][k] * l2[k][j]
            
            row.append(total)
        
        l3.append(row)

print(l3)
```
