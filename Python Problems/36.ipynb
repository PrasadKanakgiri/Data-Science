### 36. Write a program to find the compound interest

```python
"""
                A = P * (1 + r/n)**(n*t)
                
A = The future value of the investment/loan, including interest.
P = The principal investment amount (the initial sum of money).
r = The annual interest rate (decimal format, e.g., 5% = 0.05).
n = The number of times interest is compounded per year.
t = The time the money is invested or borrowed for, in years.
"""

p = float(input("Enter the principal investment amount"))
r = float(input("Enter the annual interest rate (%)")) * 0.01
n = int(input("Enter the number of times interest is compounded per year."))
t = float(input("Enter the time the money is invested or borrowed for, in years."))

A = p * (1 + r/n)**(n*t)

print(f"Compund Interest: {A:.2f}")
```
