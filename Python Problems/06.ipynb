### 6. Write a program that will tell whether the number entered by the user is odd or even.

```python
# 1st solution

number = int(input("Enter your number: "))

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# 2nd solution

number = int(input("Enter your number: "))

print("Even" if number % 2 == 0 else "Odd")
```
