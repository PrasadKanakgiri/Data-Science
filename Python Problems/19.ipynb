### 19. Write a program that will take user input of (4 digits number) and check whether the number is narcissist number or not. 

```python
# A narcissist number is same as Armstrong number

number = input("Enter your number")

total = 0


for i in number:
    total += int(i)**4

if total == int(number):
    print("Yes")
else:
    print("No")
```
