### 54. Find the index position of a particular character in another string.  

```python
# 1st solution

s = input("Enter the string: ")
char = input("Enter the character: ")

for i in range(len(s)):
    if s[i] == char:
        print("Index:", i)
        break
```

```python
# 2nd solution

s = input("Enter the string: ")
char = input("Enter the character: ")

i = s.find(char)    # returns -1 if character not present

print(i)
```

```python
# 3rd solution

s = input("Enter the string: ")
char = input("Enter the character: ")

i = s.index(char)   # causes error if character not present

print(i)
```
