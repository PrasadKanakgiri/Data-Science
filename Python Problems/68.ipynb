### 68. Write a program to merge 2 list without using the + operator

```python
# 1st solution

l1 = [2, 5, 1, 7, 4]
l2 = [9, 6, 8, 3, 10]

l1.extend(l2)

print(l1)
```

```python
# 2nd solution

l1 = [2, 5, 1, 7, 4]
l2 = [9, 6, 8, 3, 10]


for i in l2:
    l1.append(i)
    
print(l1)
```

```python
# 3rd solution

l1 = [2, 5, 1, 7, 4]
l2 = [9, 6, 8, 3, 10]

l3 = [0] * (len(l1) + len(l2)) 
j = 0

for i in range(len(l1)):
    l3[j] = l1[i]
    j = j + 1

for i in range(len(l2)):
    l3[j] = l2[i]
    j = j + 1

print(l3)
```

```python
# 4th solution

l1 = [2, 5, 1, 7, 4]
l2 = [9, 6, 8, 3, 10]


l3 = [*l1, *l2]

print(l3)
```

```python
# 5th solution

from itertools import chain

l1 = [2, 5, 1, 7, 4]
l2 = [9, 6, 8, 3, 10]

l3 = list(chain(l1, l2))

print(l3)
```
