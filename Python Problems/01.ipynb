### 1. User will input (3ages).Find the oldest one

```python
# 1st solution

n1 = int(input("Enter your age"))
n2 = int(input("Enter your age"))
n3 = int(input("Enter your age"))

oldest = max(n1, n2, n3)

if oldest == n1:
    print("Oldest is n1:", n1)
elif oldest == n2:
    print("Oldest is n2:", n2)
else:
    print("Oldest is n3:", n3)
    
```

```python
# 2nd solution

ages = []

for i in range(3):
    age = int(input("Enter your age: "))
    ages.append(age)

print("Oldest age is:", max(ages))
```

```python
# 3rd solution

ages = {
    "n1" : int(input("Enter age for n1: ")),
    "n2" : int(input("Enter age for n2: ")),
    "n3" : int(input("Enter age for n3: "))
}

oldest = max(ages, key=ages.get)
# Look at each key in ages, but compare them using the values associated with those keys.

print("Oldest is:", oldest, ":", ages[oldest])
# max() returns the key, not the value, therefore we did print(ages[oldest])
```
