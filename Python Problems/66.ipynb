### 66. Write a program to check if a list is in ascending order or not 

```python
l1 = [1, 2, 4, 5, 6, 7]

is_ascending = True

for i in range(1, len(l1)):
    if l1[i] < l1[i - 1]:
        is_ascending = False
        break

if is_ascending:
    print("Ascending Order")
else:
    print("Descending Order")
```

```python
l3 = [7, 6, 5]

if l3 == sorted(l3, reverse=True):
    print("Descending Order")
else:
    print("Ascending Order")
```
