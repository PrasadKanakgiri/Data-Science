### 34. Print first 25 prime numbers 

```python
import math

def prime(num):
    is_prime = True
    
    if num < 2:
        is_prime = False
        
    for i in range(2, int(math.sqrt(num)) + 1):
        if num % i == 0:
            is_prime = False
            break
    
    return is_prime


for i in range(1, 26):
    if prime(i):
        print(i)
    
    
        
```
