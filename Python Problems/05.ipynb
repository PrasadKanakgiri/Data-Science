## 5. Write a program that will reverse a four digit number. Also it checks whether the reverse is true.

```python
# 1st solution

num1 = input("Enter your number: ")
print("Entered number is:", num1)

num2 = reversed(num1)  # reversed() does not directly return the reversed string. It returns a reverse iterator.

num3 = ''.join(num2)

print("Reversed number is:", num3)

"""
This is what reversed does -> Give me the characters of 12345, one at a time, from the end to the beginning.

This line:
num2 = reversed(num1)

Python gets: 
5
4
3
2
1

Python doesn't create "54321".
That's what an iterator is: something that gives you items one at a time.

If you want to collect those characters into a string, you use:
num2 = ''.join(reversed(num1))

"""
```

```python
# 2nd solution

num1 = input("Enter your number: ")

l1 = list(num1)   
print("Entered number is:", ''.join(num1))

start = 0
end = len(num1) - 1

while start <= end:
    temp = num1[start]
    l1[start] = l1[end]  # this is the reason why we convereted the string to list first, we cannot reassign string as strings are immutable
    l1[end] = temp
    
    start = start + 1
    end = end - 1
    
print("Reversed number is:", ''.join(l1))
```

```python
# 3rd Solution

num1 = input("Enter your number: ")
print("Entered number is:", num1)

num2 = num1[::-1]
print("Reversed number is:", num2)
```
