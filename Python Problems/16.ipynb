### 16. Write a program that will determine weather when the value of temperature and humidity is provided by the user. 
```text
TEMPERATURE(C)      HUMIDITY(%)      WEATHER 
    >= 30              >= 90       Hot and Humid   
    >= 30              < 90             Hot
    <30                >= 90       Cool and Humid          
    <30                < 90             Cool  
```             
            
                

 



````python
temprature = int(input("Enter the temprature: "))
humidity = int(input("Enter the humidity: "))

if temprature >= 30 and humidity >= 90:
    print("Hot and Humid")
elif temprature >= 30 and humidity < 90:
    print("Hot")
elif temprature < 30 and humidity >= 90:
    print("Cool and Humid")
else:
    print("Cool")
````
