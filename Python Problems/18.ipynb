### 18. Write a program that will check whether the number is armstrong number or not. 

```python
"""
A number that equals the sum of its own digits, where each 
digit is raised to the power of the total number of digits.

153 (3 digits):
(1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153) (Yes)

372 (3 digits):
(3^3 + 7^3 + 2^3 = 27 + 343 + 8 = 378 != 372) (No)
"""

number = input("Enter your number")

length = len(number)
total = 0


for i in number:
    total += int(i)**length

if total == int(number):
    print("Yes")
else:
    print("No")
```
