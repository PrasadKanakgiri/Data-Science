### 52. Extract username from a given email.  
```text
Eg if the email is nitish24singh@gmail.com then the username 
should be nitish24singh 
```

````python
# 1st solution
s = input("Enter your username: ")

s1= ""

i = 0

while s[i] != '@':
    s1 = s1 + s[i]
    i = i + 1

print(s)
print(s1)

````

````python
# 2nd solution

s = input("Enter your username: ")

index = s.find('@')

print(s[:index])
````

````python
# 3rd solution
s = input("Enter your username: ")

username = s.split('@')[0]   # s.split('@') = ["prasad123", "gmail.com"]

print(username)
````
