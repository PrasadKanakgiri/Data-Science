### 49.  Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

```python
total = 0
avg = 0
total_nums = 0

n = int(input("Enter the num: "))

while n != 0:
    total_nums = total_nums + 1
    total = total + n
    avg = total/total_nums
    n = int(input('Another number please: '))
    
print(f"Sum = {total}, Average = {avg}")
```
