### 56. Write a program which can remove a particular character from a string.  

```python
# 1st solution

s = input("Enter the string: ")
char = input("Enter the character: ")

new_s = s.split(char)
new_s = ''.join(new_s)

print(s)
print(new_s)
```

```python
# 2nd solution

s = input("Enter the string: ")
char = input("Enter the character: ")

new_s = s.replace(char, "")
print(new_s)
```
