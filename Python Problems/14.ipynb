### 14. Hour and Minute Hands Angle

##### Given a string s representing time in 24-hour format "HH:MM", compute the smallest angle in degrees between the hour and minute hands of an analog clock.

```python
time = input("Enter your time (HH:MM): ")

hour, minutes = map(int, time.split(":"))

minute_angle = minutes * 6                          # minute angle = 360*60 = 6° per minute
hour_angle = (hour % 12) * 30 + minutes * 0.5       
#  hour % 12 to convert 24 hours format to 12 hours e.g., 23 = 23 % 12 = 11
# for evry minute hour hand moves 0.5° that is why minute * 0.5 

angle = abs(hour_angle - minute_angle)
angle = min(angle, 360 - angle)

print(f"{angle:.2f}°")
```
