### 44. Write a program to print the following pattern 
```text
1 
1 2 1 
1 2 3 2 1 
1 2 3 4 3 2 1 
1 2 3 4 5 4 3 2 1
```

````python
row = int(input('Enter number of rows: '))

for i in range(1, row + 1):
    for j in range(1, i + 1):
        print(j, end=' ')
    for k in range(i - 1, 0, -1):
        print(k, end=' ')
    print()
````
