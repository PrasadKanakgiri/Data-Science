### 23. Write a program that will swap numbers

```python
a, b = map(int, input("Enter a and b: ").split())

print(f"a = {a}, b = {b}")

a, b = b, a

print(f"a = {a}, b = {b}")
```

```python
a, b = map(int, input("Enter a and b: ").split())

print(f"a = {a}, b = {b}")

temp = a
a = b
b = temp

print(f"a = {a}, b = {b}")
```

```python
a, b = map(int, input("Enter a and b: ").split())

print(f"a = {a}, b = {b}")

a = a + b
b = a - b
a = a - b

print(f"a = {a}, b = {b}")
```
