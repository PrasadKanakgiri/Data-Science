### 32. User will provide 2 numbers you have to find the HCF of those 2 numbers 

```python
"""
The Division Method (Euclidean Algorithm)
This method is fastest for larger numbers:
1. Divide the larger number by the smaller number and find the remainder:
        (18 / 12 = 1) with a remainder of 6 
2. Take your previous divisor (12) and divide it by that remainder (6):
        (12 / 6 = 2) with a remainder of 0
3. When the remainder is 0, the last divisor you used is the HCF:
        HCF = 6
"""

a, b = map(int , input("Enter the 2 numbers: ").split())

print(a, b)


dividend = max(a, b)
divisor = min(a, b) 

while( divisor != 0):
    remainder = dividend % divisor
    
    dividend = divisor
    divisor = remainder
    

print(dividend)
```
