### 61. Write a python program to reverse a list 

```python
# 1st solution

l1 = [3, 6, 1, 4, 8]

print(l1[::-1])
```

```python
# 2nd solution

l1 = [3, 6, 1, 4, 8]

start = 0
end = len(l1) - 1

while start < end:
    # temp = l1[start] 
    # l1[start] = l1[end]
    # l1[end] = temp
    
    l1[start], l1[end] = l1[end], l1[start]
    
    start = start + 1
    end = end - 1
    
print(l1)
```

```python
# 3rd solution

l1 = [3, 6, 1, 4, 8]

l1.reverse()    # reverse() updates the original list

print(l1)
```

```python
# 4th solution

l1 = [3, 6, 1, 4, 8]

l2 = list(reversed(l1)) # reversed() creates a new list

print(l2)
```
