### 41. Write a program to print the following pattern 
```text
* 
** 
*** 
**** 
*****
```

````python
# 1st solution
n = int(input("Enter the number: "))

for i in range(n):
    for j in range(i+1):
        print("*", end="")
    print()  # next line
````

````python
# 2nd solution
n = int(input("Enter the number: "))

for i in range(1, n + 1):
    print("*" * i)
````
