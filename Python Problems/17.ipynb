### 17. Write a program that will take three digits from the user and add the square of each digit. 

```python
# 1st solution

n1, n2, n3 = map(int, input("Enter 3 numbers: ").split())

ans = n1**2 + n2**2 + n3**2

print(ans)
```

```python
# 2nd solution

n1, n2, n3 = map(int, input("Enter 3 numbers: ").split())

l1 = []

l1.extend([n1, n2, n3])
print(l1)

l1 = list(map(lambda x : x**2, l1))

print(sum(l1))
```

```python
# 3rd solution

n1, n2, n3 = map(int, input("Enter 3 numbers: ").split())

l1 = [n1, n2, n3]
total = 0

for i in l1:
    total += i**2
    
print(total)
```

```python
# 4th solution

n1, n2, n3 = map(int, input("Enter 3 numbers: ").split())

l1 = [n1, n2, n3]

squares = [x**2 for x in l1]

print(sum(squares))
```
