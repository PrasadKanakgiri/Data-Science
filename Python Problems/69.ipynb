### 69. Write a program to replace an item with a different item if found in the list

```python
# 1st solution

l1 = [1, 4, 3, 7, 8, 5, 9, 6]

replace_item = 7
replace_with = 10

for i in range(len(l1)):
    if l1[i] == replace_item:
        l1[i] = replace_with
        
print(l1)
```

```python
# 2nd solution

l1 = [1, 4, 3, 7, 8, 5, 9, 6]

replace_item = 7
replace_with = 10

if replace_item in l1:
    index = l1.index(replace_item)
    l1[index] = replace_with
    
print(l1)
```
