### 58. Write a python program to remove all the duplicates from a list 

```python
# 1st solution

l1 = [1, 2, 3, 3, 6, 6, 6, 4, 5, 5]

l2 = set(l1)

print(l2)
```

```python
# 2nd solution

l1 = [1, 2, 3, 3, 6, 6, 6, 4, 5, 5]

l2 = []

for i in l1:
    if i not in l2:
        l2.append(i)
    
print(l2)
```

```python
# 3rd solution - if u want to preserve order

l1 = [1, 2, 3, 3, 6, 6, 6, 4, 5, 5]

seen = set()

l2 = []

for i in l1:
    if i not in seen:
        seen.add(i)
        l2.append(i)
        
print(l2)
```
