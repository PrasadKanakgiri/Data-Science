### 31. Write a program to print all the unique combinations of 1,2,3 and 4 

```python
def helper(nums, current, index, result):
    if index == len(nums):
        result.append(current.copy())
        return
    
    # dont include
    helper(nums, current, index + 1, result)
    
    # include
    current.append(nums[index])
    helper(nums, current, index + 1, result)
    
    current.pop()

def combinations(nums):
    result = []
    current = []
    index = 0
    
    helper(nums, current, index, result)
    
    return result

nums = [1, 2, 3, 4]
new_nums = combinations(nums)
print(new_nums[1:])
```
