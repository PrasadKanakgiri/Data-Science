### 27. Write a program to print the first 25 odd numbers 

```python
# 1st solution

count = 25
num = 0
while count > 0:
    if num % 2 != 0:
        print(num)
        
    num = num + 1
    count = count - 1
    
```

```python
# 2nd solution

count = int(input("Enter your number: "))

num = 0
while count > 0:
    if num % 2 != 0:
        print(num)
        
    num = num + 1
    count = count - 1
```
