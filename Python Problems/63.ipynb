### 63. Write a program that can create a new list from a given list where each item in the new list is square of the item of the old list 

```python
# 1st solution

l1 = [4, 2, 6, 8, 3 , 1, 10, 9]

squares = lambda x : x**2
l2 = list(map(squares, l1))

print(l2)
```

```python
# 2nd solution

l1 = [4, 2, 6, 8, 3 , 1, 10, 9]

l2 = [i**2 for i in l1]

print(l2)
```

```python
# 3rd solution

l1 = [4, 2, 6, 8, 3 , 1, 10, 9]

l2 = []

for i in l1:
    l2.append(i**2)
    
print(l2)
```
