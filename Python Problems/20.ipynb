### 20. Write a program that will give you the in hand salary after deduction of 
```text
HRA(10%),DA(5%),PF(3%), and tax(if salary is between 5-10 lakh–10%),(11-20lakh–20%),(20<– 30%)(0-1lakh print k).
```

````python
salary = float(input("Enter your annual salary: "))

hra = salary * 0.10
da = salary * 0.05
pf = salary * 0.03

tax = 0

if salary <= 100000:
    print("K")
else:
    if salary > 500000:
        tax += (min(salary, 1000000) - 500000) * 0.10   # - 500000 because no tax for 1L-5L
    
    if salary > 1000000:
        tax += (min(salary, 2000000) - 1000000) * 0.20  # - 1000000 because 5L-10L is already calculated in above if condition
        
    if salary > 2000000:
        tax += (salary - 2000000) * 0.30    # - 2000000 because 10L-20L is already calculated in above if condition

in_hand = salary - hra - da - pf - tax

print("HRA:", hra)
print("DA:", da)
print("PF:", pf)
print("Tax:", tax)
print("In-hand salary:", in_hand)
````
