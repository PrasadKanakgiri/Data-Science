### 73. Write a program that can convert an integer to string. 

```python
# 1st solution

i = 9837
print(type(i))

s = str(i)
print(type(s))
```

```python
# 2nd solution

i = 9

s = chr(i + 48)  
'''
'0' → 48, 
'9' -> 57 
i + 48 = 9 + 48 = 57

Now 57 is ascii value for '9'
'''
print(type(s))
```

```python
# 3rd solution

i = 12345
s = ""

while i > 0:
    digit = i % 10
    s = chr(digit + 48) + s
    i = i // 10
    
print(s)
print(type(s))
```

```python
# 4th solution

i = -12345

if i == 0:
    s = "0"
else:
    negetive = False
    
    if i < 0:
        negetive = True
        i = -i
        
        s = ""
        
        while i > 0:
            digit = i % 10
            s = chr(digit + 48) + s
            i = i // 10
            
        if negetive:
            s = '-' + s

print(s)
print(type(s))
```
