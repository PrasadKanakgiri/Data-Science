### 59. Write a python program to convert a string to title case without using the title() 

```python
# 1st solution

s = "hi how are you"

l = s.split()

print(l)

for i in range(len(l)):
    l[i] = l[i].capitalize()

print(l)
  
s = ' '.join(l)

print(s)
```

```python
# 2nd solution

s = "hi how are you"

l = [i.capitalize() for i in s.split()]

new_s = " ".join(l)

print(new_s)
```

```python
# 3rd solution

s = "hi how are you"

l = s.split()

new_s = ""

for i in l:
    new_s = new_s + i.capitalize() + " "

print(new_s)
```
