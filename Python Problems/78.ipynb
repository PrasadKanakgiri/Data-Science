### 78. Write a program that can find the most used word in a bollywood song 

```python
song = "tum hi ho tum hi ho zindagi ab tum hi ho"

words = song.lower().split()

count = {}

for word in words:
    
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

# most_used = max(count, key=count.get)

most_used = ""
highest_count = 0

for word in count:
    if count[word] > highest_count:
        highest_count = count[word]
        most_used = word


print(most_used)
print(highest_count)
```
