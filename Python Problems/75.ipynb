### 75. Write a program that can check if you can perform matrix multiplication on 2 matrices 

```python
l1 = [
    [1, 2, 3, 4, 5],
    [5, 6, 7, 8, 9]
]

l2 = [
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
]

r1 = len(l1)
c1 = len(l1[0])

r2 = len(l2)
c2 = len(l2[0])

if c1 == r2:
    print("Matrix multiplication possible")
else:
    print("Matrix multiplication not possible")
```
