### 8. Write a program to find the euclidean distance between two coordinates

```python
# 1st solution

# 2D space: (d = sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2})

import math

x1 = float(input("Enter value of x for your fist coordinate"))
y1 = float(input("Enter value of y for your fist coordinate"))
x2 = float(input("Enter value of x for your second coordinate"))
y2 = float(input("Enter value of y for your second coordinate"))

d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print(f"Euclidean distance between ({x1}, {y1}) and ({x2, y2}) is {d:.1f}")
```

```python
# 2nd solution

import math

x1 = float(input("Enter x for first coordinate: "))
y1 = float(input("Enter y for first coordinate: "))
x2 = float(input("Enter x for second coordinate: "))
y2 = float(input("Enter y for second coordinate: "))

d = math.dist((x1, y1), (x2, y2))

print(f"Euclidean distance between ({x1}, {y1}) and ({x2}, {y2}) is {d}")
```
