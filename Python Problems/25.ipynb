### 25. Write a program that can multiply 2 numbers provided by the user without using the * operator 

```python
a, b = map(int, input("Enter a and b: ").split())

print("a = ", a)
print("b = ", b)

total = 0

remaining = abs(b)

while remaining > 0:
    total = total + a
    remaining = remaining - 1
    
if b < 0:
    total = -total
    
print(f"{a} * {b} = {total}")
```
