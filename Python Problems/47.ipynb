### 47. 47. Write a Python Program to Find the Sum of the Series till the nth term:  
```text
1 + 2^2/2 + 3^3/3 + … n^n/n 
n will be provided by the user 
```

````python
def total_sum(n):
    if n == 1:
        return 1
    
    return (n**n/n) + total_sum(n-1)

n = int(input("Enter the value of n: "))

ans = total_sum(n)
print(ans)
````
