### 48. The natural logarithm can be approximated by the following series. If x is input through the keyboard, write a program to calculate the sum of the first n terms of this series. 
```text
(x-1/x) + (1/2)(x-1/x)**2 + (1/2)(x-1/x)**3 + (1/2)(x-1/x)**4 + (1/2)(x-1/x)**n
```

````python
def total_sum(x, n):
    total = (x - 1)/ x
    
    for i in range(2, n+1):
        total = total + (1/2)*((x - 1)/ x)**i
    
    return total

ans = total_sum(5, 7)
print(ans)
````
