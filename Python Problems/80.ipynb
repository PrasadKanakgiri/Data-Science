### 80. Write a program to merge two given dictionary 

```python
# 1st solution

d1 = {
    "a": 10,
    "b": 20,
    "c": 30
}

d2 = {
    "d": 40,
    "e": 50,
    "f": 60
}

d1.update(d2)

print(d1)
```

```python
# 2nd solution

d1 = {
    "a": 10,
    "b": 20,
    "c": 30
}

d2 = {
    "d": 40,
    "e": 50,
    "f": 60
}

d3 = {}

for key in d1:
    d3[key] = d1[key]
    
for key in d2:
    d3[key] = d2[key]
    
print(d3)
```

```python
# 3rd solution

d1 = {
    "a": 10,
    "b": 20,
    "c": 30
}

d2 = {
    "d": 40,
    "e": 50,
    "f": 60
}

d3 = {**d1, **d2}

print(d3)
```

```python
# 4th solution

d1 = {
    "a": 10,
    "b": 20,
    "c": 30
}

d2 = {
    "d": 40,
    "e": 50,
    "f": 60
}

d3 = d1 | d2

print(d3)
```
