### 28. Write a program to print whether a given number is prime number or not 

```python
# 1st solution

# A number is prime if it is divisible by more than 1 numbers

num = int(input("Enter the number: "))

total = 0
    
for i in range(1, num + 1):
    if num % i == 0:
        total = total + 1

if total == 2:            # A prime number has exactly 2 divisors: 1 and itself.
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
# 2nd solution

# Instead of counting all divisors, we can assume the number is prime and look for a number that divides it.

num = int(input("Enter the number: "))

is_prime = True

if num < 2:
    is_prime = False
    
for i in range(2, num):
    
    if num % i == 0:
        is_prime = False
        break
    
if is_prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
# 3rd solution

# If a number has a divisor, one of its factor pairs will be <= √n.
# So we only need to check divisors up to √n; after that, factors repeat.

import math
num = int(input("Enter the number: "))

is_prime = True

if num < 2:
    is_prime = False

for i in range(2, int(math.sqrt(num))):     # int(math.sqrt(num)) because math.sqrt(num) returns float
    if num % i == 0:
        is_prime = False
        break

if is_prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```
