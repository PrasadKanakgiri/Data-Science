### 29. Print all the armstrong numbers in the range of 100 to 1000

```python
def is_armstrong(num):
    num = str(num)
    total = 0
    
    for i in num:
        total = total + (int(i) ** len(num))
        
    if total == int(num):
        return True
    
    return False

for i in range(100, 1001):
    if is_armstrong(i):
        print(i)
```
