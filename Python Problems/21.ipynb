### 21. Write a menu driven program 
```text
1. cm to ft  
2. kl to miles  
3. usd to inr  
4. exit 
```

````python
# 1st solution

while True:
    num = input("""
                Choose a number:
                1. cm to ft
                2. kms to miles
                3. USD to INR
                4. Exit
                """)
    if num == "1":
        cms = float(input("Enter the value (cms): "))
        feet = cms / 30.48
        print(f"{cms} cms = {feet:.1f} ft")  
           
    elif num == "2":
        kms = int(input("Enter the value (kms): "))
        miles = kms * 0.62
        print(f"{kms} kms = {miles:.1f} miles")
        
    elif num == "3":
        usd = int(input("Enter the value (USD): "))
        inr = usd * 95.12
        print(f"{usd} USD = {inr:.1f} INR")
        
    elif num == "4":
        print("Exiting...")
        break
    
    else:
        print("Invalid choice. Please select 1-4.")
````

````python
# 2nd solution

def cm_to_ft():
    cms = float(input("Enter the value (cms): "))
    feet = cms / 30.48
    print(f"{cms} cms = {feet:.1f} ft")  

def km_to_mile():
    kms = float(input("Enter the value (kms): "))
    miles = kms * 0.62
    print(f"{kms} kms = {miles:.1f} miles")

def usd_to_inr():
    usd = float(input("Enter the value (USD): "))
    inr = usd * 95.12
    print(f"{usd} USD = {inr:.1f} INR")


while True:
    num = input("""
                Choose a number:
                1. cm to ft
                2. kms to miles
                3. USD to INR
                4. Exit
                """)
    if num == "1":
        cm_to_ft()
        
    elif num == "2":
        km_to_mile()
        
    elif num == "3":
        usd_to_inr()
        
    elif num == "4":
        print("Exiting...")
        break
    
    else:
        print("Invalid choice. Please select 1-4.")
````
