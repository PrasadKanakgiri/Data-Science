### 64. Write a program that can reverse words of a given string. 
```text
e.g., 
Input: Hello how are you 
Output: you are how Hello 
```

````python
# 1st solution

s1 = "Hello how are you"

l1 = s1.split()
print(l1)
l2 = l1[::-1]
print(l2)
s2 = " ".join(l1[::-1])

print(s2)
````

````python
# 2nd solution

s1 = "Hello how are you"

l1 = s1.split()
l2 = []

for i in range(len(l1) - 1, -1, -1):
    l2.append(l1[i])

s2 = " ".join(l2)

print(s2)
````

````python
# 3rd solution

s1 = "Hello how are you"

l1 = s1.split()

s2 = " ".join(reversed(l1))

print(s2)
````

````python
# 4th solution

s1 = "Hello how are you"

s2 = " ".join(reversed(s1.split()))

print(s2)
````
