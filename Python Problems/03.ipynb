## 3. User will input (2numbers).Write a program to swap the numbers

```python
# 1st solution

num1 = int(input("Enter the 1st number: "))
num2 = int(input("Enter the 2nd number: "))

print("Before swapping n1 =", num1, "n2 =", num2)

temp = num1
num1 = num2
num2 = temp

print("After swapping n1 =", num1, "n2 =", num2)
```

```python
# 2nd solution

num1 = int(input("Enter the 1st number: "))
num2 = int(input("Enter the 2nd number: "))

print("Before swapping n1 =", num1, "n2 =", num2)

num1, num2 = num2, num1
# Python evaluates the right side first, then assigns the values to the variables on the left.
"""
If entered number are num1 = 10, num2 = 20
Step 1: Right side is evaluated first which is num2, num1 => num2 = 20, num1 = 10
Step 2: Assigns the values to the variables on the left => num1 = 20, num2 = 10
"""


print("After swapping n1 =", num1, "n2 =", num2)
```
