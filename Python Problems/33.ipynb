### 33. User will provide 2 numbers you have to find the by LCM of those 2 numbers 

```python
# 1st solution
"""
Using multiples
For two numbers, generate multiples of the larger number and check which one is divisible by the smaller number.
For example, 12 and 18:
18 x 1 = 18  → not divisible by 12
18 x 2 = 36  → divisible by 12
So:
LCM = 36
"""

a, b = map(int, input("Enter the 2 numbers: ").split())

print(a, b)

num1 = max(a, b)
num2 = min(a, b)
multiply = 1
while (num1 * multiply) % num2 != 0:
    multiply = multiply + 1

lcm = num1 * multiply
print(lcm)
```

```python
# 2nd solution
"""
LCM x HCF = product of 2 numbers
"""

def hcf(a, b):
    dividend = max(a, b)
    divisor = min(a, b)

    while divisor != 0:
        remainder = dividend % divisor

        dividend = divisor
        divisor = remainder

    return dividend


a, b = map(int, input("Enter the 2 numbers: ").split())

print(a, b)

lcm = (a * b) // hcf(a, b)

print(lcm)
```
