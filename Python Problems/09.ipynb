### 9. Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

```python
'''
# Whenever you want multiple integer inputs in one line, use: 
a, b, c = map(int, input().split())
'''

angle1, angle2, angle3 = map(int, input("Enter you 3 angles: ").split())

print(angle1)
print(angle2)
print(angle3)

total = angle1 + angle2 + angle3

if total == 180 and angle1 > 0 and angle2 > 0 and angle3 > 0:
    print("Entered angles can form a triangle")
else:
    print("Entered angles cannot form a triangle")
```
