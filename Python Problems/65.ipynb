### 65. Write a program that can count the number of words in a given string 

```python
# 1st solution

s1 = "Hi how are you"

l1 = s1.split()

length = len(l1)

print(length)
```

```python
# 2nd solution

s1 = "Hi how are you"

length = len(s1.split())

print(length)
```

```python
# 3rd solution

s1 = "Hi how are you"

length = 0

for i in s1.split():
    length = length + 1
    
print(length)
```
