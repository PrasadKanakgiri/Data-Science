### 74. Write a program to print the shape of a matrix.

```python
# 1st solution

l = [
    [1, 2, 3, 4, 5],
    [5, 6, 7, 8, 9]
]

r = 0
c = 0

for i in l:
    r = r + 1
    
c = len(l[0])
print(f"{r} x {c}")
```

```python
# 2nd solution

l = [
    [1, 2, 3, 4, 5],
    [5, 6, 7, 8, 9]
]

r = len(l)
c = len(l[0])

print(f"{r} x {c}")
```
