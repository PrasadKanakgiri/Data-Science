### 46. Write a program to calculate the sum of the following series till the 
```text
nth term 
1/1! + 2/2! + 3/3! + 4/4! +…….+ n/n! 
n will be provided by the user
```

````python
def factorial(n):
    if n <= 1:
        return 1
    
    return n * factorial(n - 1)

def total_sum(n):
    if n == 1:
        return 1
    
    return (n/factorial(n)) + total_sum(n - 1)

n = int(input("Enter the value of n: "))

ans = total_sum(n)
print(f"{ans:.2f}")
````
