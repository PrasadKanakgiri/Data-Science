### 50. Write a program that accepts 2 numbers from the user a numerator and a denominator and then simplifies it 
```text
e.g., if the num = 5, den = 15 the answer should be ⅓ 
e.g., if the num = 6, den = 9 the answer should be ⅔ 
```

````python
# 1st solution

def HCF(num, den):
    dividend = max(num, den)
    divisor = min(num, den)
    
    while divisor != 0:
        remainder = dividend % divisor
        
        dividend = divisor
        divisor = remainder
    
    return dividend

hcf = HCF(6, 9)

numerator = 6//hcf
denominator = 9//hcf

print(f"{numerator}/{denominator}")
````

````python
# 2nd solution
class Fraction:
    def __init__(self, num, den):
        self.num = num
        self.den = den
    
    def __str__(self):
        
        num, den = self.simplify(self.num, self.den)

        return f"{num}/{den}"
    
    def simplify(self, num, den):
        hcf = self.HCF(num, den)
        
        return num // hcf, den // hcf
        
    def HCF(self, num, den):
        dividend = max(num, den)
        divisor = min(num, den)
        
        while divisor != 0:
            remainder = dividend % divisor
            
            dividend = divisor
            divisor = remainder
        
        return dividend
    
fraction = Fraction(6, 9)

print(fraction)
````
