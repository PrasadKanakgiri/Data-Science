### 57. Write a program that can check whether a given string is palindrome or not. 

```python
# 1st solution

s = input("Enter the string: ")

s1 = ''.join(reversed(s))


# print(s)
# print(s2)

if s == s1:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 2nd solution

s = input("Enter the string: ")


if s == s[::-1]:
        print("Palindrome")
else:
    print("Not Palindrome")
```
