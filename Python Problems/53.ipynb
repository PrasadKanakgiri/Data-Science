### 53. Count the frequency of a particular character in a provided string. 
```text
e.g., 'hello how are you' is the string, the frequency of h in this string is 2.
```

````python
# 1st solution

s = "hello how arhe yHou"

count = 0

for i in s:
    if i == 'h' or i == 'H':
        count = count + 1

print(count)
````

````python
# 2nd solution

s = "hello how arhe yHou"

count = 0

for i in s:
    if i.lower() == 'h':
        count = count + 1

print(count)
````

````python
# 3rd solution

s = "hello how arhe yHou"

count = s.lower().count('h')

print(count)
````
