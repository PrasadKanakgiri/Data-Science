### 71. Write a program that can perform union and intersection on 2 given list.

```python
# 1st solution
l1 = [1, 2, 3, 4, 5, 6]
l2 = [7, 8, 9, 3, 4, 1]

unionArray = set(l1) | set(l2)
intersectionArray = set(l1) & set(l2)

print(unionArray)
print(intersectionArray)
```

```python
# 2nd solution

l1 = [1, 2, 3, 4, 5, 6]
l2 = [7, 8, 9, 3, 4, 1]

unionArray = []
intersectionArray = []

for i in l1:
    if i not in unionArray:
        unionArray.append(i)

    if i in l2:
        intersectionArray.append(i)
        
for i in l2:
    if i not in unionArray:
        unionArray.append(i)
        
print(unionArray)
print(intersectionArray)

```

```python
# 3rd solution

l1 = [1, 2, 3, 4, 5, 6]
l2 = [7, 8, 9, 3, 4, 1]

s1 = set(l1)
s2 = set(l2)

unionArray = s1.union(s2)
intersectionArray = s1.intersection(s2)

print(unionArray)
print(intersectionArray)
```
