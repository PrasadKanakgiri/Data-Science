### 37. Write a Python program that accepts an integer (n) and computes the value of n+nn+nnn. 

```python
n = 8
total = n
multiple = 10
l1 = []
num = n
l1.append(num)

for i in range(2):
    num = num + (n*multiple)
    total = total + num
    l1.append(num)
    multiple = multiple*10

print(l1)
print(total)
        
```
