## 4. Write a program that will give you the sum of 3 digits

```python
# 1st solution

total = 0

for i in range(3):
    num = int(input("Enter your number: "))
    total += num

print("Sum is:", total)
```

```python
# 2nd solution

numbers = []

for i in range(3):
    num = int(input("Enter your number: "))
    numbers.append(num)

print("Sum is:", sum(numbers))
```
