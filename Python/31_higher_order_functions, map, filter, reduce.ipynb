````python
def return_sum(func, L):
    result = 0
    
    for i in L:
        if func(i) == True:
            result = result + i
    
    return result

L = [10, 4, 3, 11, 5, 9, 13, 27, 15, 8, 14]

even = lambda x : x % 2 == 0
odd = lambda x : x % 2 != 0
div3 = lambda x : x % 3 == 0

print(return_sum(even, L))
print(return_sum(odd, L))
print(return_sum(div3, L))

````

# map() in Python

`map()` is a built-in function used to **apply the same function to every element of an iterable such as a list**.

### Syntax
```python
map(function, iterable)

### Example

Suppose we want to square every number:

```python
nums = [1, 2, 3, 4, 5]

result = map(lambda x: x * x, nums)

print(list(result))
```

````python
nums = [1, 2, 3, 4, 5]

result1 = map(lambda x: x * x, nums)

print(result1)
print(list(result1))
````

# filter() in Python

`filter()` is used to select elements from an iterable based on a condition.

### Syntax
```python
filter(function, iterable)
```
**The function should return True or False.**

True → keep the element
False → remove the element

### Example: Get even numbers
```python
nums = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, nums)

print(list(result))
```
### Output:

[2, 4, 6]

# `map()` vs `filter()`

This distinction is very important:

**map()    → TRANSFORM every element**

**filter() → SELECT elements**

### Example:
```python
nums = [1, 2, 3, 4, 5]
```
### map → transform
list(map(lambda x: x * 2, nums))
 [2, 4, 6, 8, 10]

### filter → select
list(filter(lambda x: x % 2 == 0, nums))
 [2, 4]

### So you can remember:

**`map()` changes the values, while filter() chooses which values to keep.**

````python
nums1 = [1, 2, 3, 4, 5]

result1 = filter(lambda x: x % 2 == 0, nums1)

print(result1)
print(list(result1))

# same with map
nums2 = [1, 2, 3, 4, 5]

result2 = map(lambda x: x % 2 == 0, nums2)

print(result2)
print(list(result2))
````

````python
nums = [1, 2, 3, 4, 5]

result = map(lambda x: x % 2 == 0, nums)

print(result)
print(list(result))
````

# `reduce()` in Python

`reduce()` is used to combine all elements of an iterable into a single value.

**It is available in the functools module.**

### Syntax
```python
from functools import reduce

reduce(function, iterable)
```
### Example: Sum of numbers
```python
from functools import reduce

nums = [1, 2, 3, 4, 5]

result = reduce(lambda a, b: a + b, nums)

print(result)
```
### Output:

15

### How does it work?

`reduce()` takes two elements at a time and keeps combining them.

[1, 2, 3, 4, 5]

1 + 2 = 3

3 + 3 = 6

6 + 4 = 10

10 + 5 = 15

# A simple way to remember:

**Map = change, Filter = choose, Reduce = combine.**

````python
from functools import  reduce

nums = [1, 2, 3, 4, 5]

result1 = reduce(lambda x, y : x + y, nums)
print(result1)

result2 = reduce(lambda x, y : x if x > y else y, nums)
print(result2)
````
