### String Indexing

Indexing is used to **access individual characters** of a string.

The index starts from **0**.

```python
name = "Python"

print(name[0])
print(name[1])
print(name[2])
print(name[3])
print(name[4])
print(name[5])
```

### Negative Indexing

Negative indexing is used to access characters from the end of the string.

The last character has an index of -1.

```python
name = "Python"

print(name[-1])
print(name[-2])
print(name[-3])
```
### Output:

n
o
h

````python
name = "Python"

print(name[1])
print(name[-1])
````

### String Slicing

Slicing is used to **extract a part of a string**.

### Syntax

```python
string[start:stop:step]
```
`start` → Index from where slicing starts.

`stop` → Index where slicing stops. The stop index is not included.

`step` → Number of positions to move. Default is 1.

### Example
```python
name = "Python"

print(name[0:3])
```
### Output
Pyt

Here, indexing starts from `0` and stops before index `3`.

### Using Step
```python
name = "Python"

print(name[0:6:2])
```
### Output:

Pto

Here, the step is `2`, so every second character is selected.

### Negative Slicing

Slicing can also be done using negative indexes.
```python
name = "Python"

print(name[-3:])
```
### Output:

hon

### Reverse a String

A string can be reversed using a negative step.
```pyhton
name = "Python"

print(name[::-1])
```
### Output:

nohtyP
