# input() Function

The `input()` function is used to **take input from the user**.

The value entered by the user is stored in a variable.

```python
name = input("Enter your name: ")
print(name)
```
Here, the user enters their name, and it is stored in the `name` variable.

### Important

The `input()` function **always returns the input as a string (`str`)**, even if the user enters a number.

```python
age = input("Enter your age: ")

print(age)
print(type(age))
```

If the user enters `25`, the output will be:

25
<class 'str'>

````python
a = input("First Number")
b = input("Second Number")
````

# type()

The `type()` function is used to **check the data type of a value or variable**.

```python
a = 10
print(type(a))
```

Here, a contains an integer, so type(a) returns `int`.

````python
print(type(10))
print(type(3.5))
print(type(True))
print(type(5 + 4j))
print(type("India"))
print(type([1, 2, 3]))
````
