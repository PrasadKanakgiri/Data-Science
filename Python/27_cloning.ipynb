# Shallow Copy vs Deep Copy

## 1. Shallow Copy

A **shallow copy** creates a new outer object, but the **nested objects are still shared**.

### Example

```python
a = [[1, 2], [3, 4]]

b = a.copy()

print(a is b)
```
#### Output:

False

`a` and `b` are different outer lists:



The outer list is copied, but the inner lists are shared.

#### Proof
```python
a = [[1, 2], [3, 4]]
b = a.copy()

b[0].append(100)

print(a)
print(b)
```
#### Output:

[[1, 2, 100], [3, 4]]
[[1, 2, 100], [3, 4]]

Why did `a` also change?

Because:



Both a[0] and b[0] refer to the same inner list.

## 2. Deep Copy

A deep copy creates a new outer object and also creates copies of nested objects.
```python
import copy

a = [[1, 2], [3, 4]]

b = copy.deepcopy(a)
```
Conceptually:



Here, the inner lists are also separate objects.

#### Proof
```python
import copy

a = [[1, 2], [3, 4]]

b = copy.deepcopy(a)

b[0].append(100)

print(a)
print(b)
```
#### Output:

[[1, 2], [3, 4]]
[[1, 2, 100], [3, 4]]

Changing `b` does not affect `a`.

````python
# Shallow Copy
print("----------------Shallow Copy----------------")
print("----Normal List----")
a = [1, 2, 3, 4]
b = a.copy()

b.append(5)

print(a)
print(b)

print("----Nested List----")
a = [[1, 2, 3, 4], [5, 6, 7, 8]]
b = a.copy()

b[0].append(10)

print(a)
print(b)

print("----------------Deep Copy----------------")
print("----Nested List----")

import copy
a = [[1, 2, 3, 4], [5, 6, 7, 8]]
b = copy.deepcopy(a)

b[0].append(10)

print(a)
print(b)

````

# Cloning

### Definition

**Cloning means creating a new object with the same contents as an existing object.**

The important point is:

> The cloned object is a **separate object** from the original.

---

## Example: List Cloning

```python
a = [10, 20, 30]

b = a.copy()

print(a)
print(b)

print(a is b)
```
Output:

[10, 20, 30]
[10, 20, 30]
False

**Although both lists have the same contents, they are different objects.**

a ─────► [10, 20, 30]

b ─────► [10, 20, 30]

There are two separate list objects.

#### Why Cloning Is Useful

Consider:
```python
a = [10, 20, 30]
b = a

b.append(40)

print(a)
```
#### Output:

[10, 20, 30, 40]

#### Why?

**Because `a` and `b` refer to the same list:**



#### With Cloning
```python
a = [10, 20, 30]

b = a.copy()

b.append(40)

print(a)
print(b)
```
#### Output:

[10, 20, 30]
[10, 20, 30, 40]

#### Now:

a ─────► [10, 20, 30]

b ─────► [10, 20, 30, 40]

**The original list is not affected.**

Ways to Clone a List
1. Using copy(): 
b = a.copy()

2. Using Slicing: 
b = a[:]

3. Using list(): 
b = list(a)

**All three create a new list object.**

### Shallow Cloning

The methods above create a shallow copy.

For a simple list:
```python
a = [10, 20, 30]
b = a.copy()
```
This is straightforward:

a ─────► [10, 20, 30]

b ─────► [10, 20, 30]

#### But with nested lists, there is an important difference.
```python
a = [[1, 2], [3, 4]]
b = a.copy()
```
#### Conceptually:



The outer list is new, but the nested lists are still shared.

Therefore, this:
```python
b[0].append(100)
```
can also affect `a`.

### Deep Cloning

A deep copy creates independent copies of nested objects as well.
```python
import copy

a = [[1, 2], [3, 4]]

b = copy.deepcopy(a)
```
Conceptually:



**Now the nested lists are also separate objects.**

````python
a = [1, 2]
b = [3, 4]
c = (a, b)

print(c)
print(a)
c[0][0] = 100

print(c)
print(a)
````
