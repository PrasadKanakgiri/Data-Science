# Built-in Functions

Built-in functions are **functions that are already provided by Python**.

We can use them directly without creating or defining them ourselves.

### Common Built-in Functions

| Function | Purpose | Example |
|---|---|---|
| `print()` | Displays output | `print("Hello")` |
| `input()` | Takes input from the user | `input("Enter name: ")` |
| `type()` | Checks the data type | `type(10)` |
| `int()` | Converts a value to an integer | `int("10")` |
| `abs()` | Returns the absolute value | `abs(-10)` |
| `pow()` | Returns a number raised to a power | `pow(2, 3)` |
| `min()` | Returns the smallest value | `min(10, 20, 5)` |
| `max()` | Returns the largest value | `max(10, 20, 5)` |
| `round()` | Rounds a number | `round(3.14)` |
| `divmod()` | Returns quotient and remainder | `divmod(10, 3)` |
| `bin()` | Converts a number to binary | `bin(10)` |
| `oct()` | Converts a number to octal | `oct(10)` |
| `hex()` | Converts a number to hexadecimal | `hex(10)` |
| `id()` | Returns the unique identity of an object | `id(10)` |
| `ord()` | Returns the Unicode value of a character | `ord("A")` |
| `len()` | Returns the length of an object | `len("Hello")` |
| `sum()` | Returns the sum of values | `sum([1, 2, 3])` |
| `help()` | Provides information about Python objects | `help(print)` |
