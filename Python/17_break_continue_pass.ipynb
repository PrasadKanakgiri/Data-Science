# break, continue and pass

`break`, `continue`, and `pass` are used to **control the flow of loops and code execution**.

### break

`break` is used to **stop the loop immediately**.

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```
### Output:

1
2

When `i` becomes `3`, the loop stops.

### continue

`continue` is used to **skip the current iteration** of a loop and move to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```
### Output:

1
2
4
5

When `i` becomes `3`, the `continue` statement skips that iteration and the loop moves to the next value.

### pass

`pass` is used when you want to **leave a block of code empty**.

It does nothing when executed.

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```
### Output:

1
2
3
4
5

Here, `pass` does nothing, so the loop continues normally.
