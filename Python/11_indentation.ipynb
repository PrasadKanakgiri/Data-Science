# Indentation

Indentation means **adding spaces at the beginning of a line**.

Python uses indentation to **define blocks of code**.

Unlike some other programming languages, Python does not use `{}` to define a block of code.

### Example

```python
age = 18

if age >= 18:
    print("You are eligible to vote")
```
Here, the print() statement is indented, so it belongs to the if block.

### Incorrect Indentation
```python
age = 18

if age >= 18:
print("You are eligible to vote")
```
This will give an IndentationError because the print() statement is not indented.
