# Functions

### Definition

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same code repeatedly, we can define a function once and **call it whenever needed**.

---

## Why Use Functions?

Without a function:

```python
print("Hello")
print("Hello")
print("Hello")
```

### With a function:
```python
def greet():
    print("Hello")

greet()
greet()
greet()
```
#### Output:

Hello
Hello
Hello

```python
Let's use your example:

def is_even(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

is_even(9)
```
### 1. When Python reads the function definition

When Python executes:
```python
def is_even(num):
    ...
```
Python creates a **function object** in memory.

The name `is_even` refers to that function object.

#### Conceptually:



The function body does not execute yet.

### 2. Then you call the function

When Python reaches:
```python
is_even(9)
```
Python creates a new function **execution frame**.

**Think of a frame as a temporary workspace created for that particular function call.**



The argument `9` is assigned to the parameter `num`.

So:
```python
is_even(9)
```
effectively gives:

num ─────► 9

inside that function call.

### 3. The function executes

Now Python executes:
```python
if num % 2 == 0:
```
Since:

num → 9

Python evaluates:

9 % 2

which gives:

1

Therefore:
```python
else:
    print("Odd")
```
runs.

### 4. What happens when the function finishes?

After:
```
print("Odd")
```
there is nothing else to execute.

So the function returns.

**The function's execution frame is destroyed/removed because it is no longer needed.**

Conceptually:



The **function object** still exists because the name `is_even` still refers to it.



# `print()` vs `return`
### Using print
```python
def add (a, b):
    print(a + b)

result = add(10, 20)

print(result)
```
### Output:

30
None

#### Why?

add(10, 20)

     ↓

  print(30)

     ↓

  displays 30

     ↓

function ends

     ↓

no return statement

     ↓

Python returns None

     ↓

result = None

**So `print()` only displays the value. It does not send the value back to the caller.**

### Using return
```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```
#### Output:

30

Here:

add(10, 20)

     ↓

  10 + 20

     ↓

    30

     ↓

 return 30

     ↓

result = 30

**So `return` sends a value back to the place where the function was called.**

````python
def is_even(num):
    """
    This function tells whether a given number is even or odd
    Input - any valid integer
    output - odd/even
    """
    if type(num) == int:
        if num % 2 == 0:
            return "Even"
        else:
            return "Odd"
    else:
        return "Enter a valid integer"

print(is_even(2))

print(is_even.__doc__)  # print the doc string
# print(print.__doc__)
````

# Arbitary Argument

````python
def add(*numbers):
    sum = 0
    print(numbers)
    print(type(numbers))
    for i in numbers:
        sum = sum + i
    return sum
    
print(add(1,2,3,4,5))
print(add(1,2,3))
````

````python
'''
Because x is assigned inside the function, Python treats it as a local variable, 
so print(x) tries to access the local x before it has been assigned a value, 
causing an UnboundLocalError.
'''

def g(y):
    print(x)
    x = x + 1  # this will throw error
    # You can use a global varaible but you cannot change it without using 'global' inside the function
    print(x)

x = 5
g(x)
print(x)
````

````python
def g(y):
    global x
    print(x)
    x = x + 1
    print(x)

x = 5
g(x)
print(x)
````

````python
def f(x):
    x = x + 1
    print("In f(x): x = ", x)
    return x

x = 3
z = f(x)

print("In main program scope: z = ", z)
print("In main program scope: x = ", x)
````

# Nested Functions

````python
def f():
    print("Inside f")
    
    def g():
        print("Inside g")
    
    g()
        
print(f())
````

````python
def g(x):
    def h():
        x = 'abc'
        # print("in h(x): x = ", x)
    
    x = x + 1
    print("in g(x): x = ", x)
    h()
    return x

x = 3
z = g(x)

print("In main program scope: z = ", z)
print("In main program scope: x = ", x)
````

# Functions are objects too

**A function will behave just like how integers, strings and other data type in python do.**

````python
def f(num):
    return num*num

print(f(2))
x = f # this is possible in python because everything in python is an object
print(x(2))

del f # this is possible in python as it just an object
# f(3) -> will throw error as f is deleted

# you can store the funtion in a list  
l1 = [1, 2, 3, x]
print(l1)

print(l1[-1](3))

l2 = [1, 2, 3, x(4)]
print(l2)
````

````python
def f():
    def x(a, b):
        return a + b
    return x

val = f()(3, 4)
print(val)
````
