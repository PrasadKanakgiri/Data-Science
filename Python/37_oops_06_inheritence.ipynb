# Inheritance in Python

**Inheritance** is an OOP concept where a **child class gets the properties and methods of a parent class**.

In simple words:

> **Inheritance allows one class to reuse the code of another class.**

It represents an **IS-A relationship**.

For example:

```text
Car IS-A Vehicle
Dog IS-A Animal
Manager IS-A Employee
```

---

# Why Do We Need Inheritance?

Suppose we have:

```python
class Vehicle:

    def start(self):
        print("Vehicle started")

    def stop(self):
        print("Vehicle stopped")
```

Now suppose we create a `Car` class.

Without inheritance, we would have to write the same methods again:

```python
class Car:

    def start(self):
        print("Vehicle started")

    def stop(self):
        print("Vehicle stopped")

    def drive(self):
        print("Car is driving")
```

This creates **duplicate code**.

Instead, we can inherit from `Vehicle`.

```python
class Car(Vehicle):

    def drive(self):
        print("Car is driving")
```

Now `Car` automatically gets the methods of `Vehicle`.

---

# Basic Syntax

```python
class Parent:

    # properties and methods
    pass


class Child(Parent):

    # additional properties and methods
    pass
```

Here:

```text
Parent
  ↑
  |
Child
```

The `Child` class inherits from the `Parent` class.

---

# Example

```python
class Vehicle:

    def start(self):
        print("Vehicle started")

    def stop(self):
        print("Vehicle stopped")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


car1 = Car()

car1.start()
car1.stop()
car1.drive()
```

Output:

```text
Vehicle started

Car is driving
```

Notice that we never defined `start()` or `stop()` inside `Car`.

But `Car` can still use them because:

```python
class Car(Vehicle):
```

means:

> `Car` inherits from `Vehicle`.

Balance: 4000
```

`SbiAtm` inherited:

```text
Atm
│
├── balance
├── check_balance()
└── deposit()
```

and added its own:

```text
SbiAtm
│
└── withdraw()
```

* Instance methods
* Class methods
* Class variables
* Other accessible attributes

For example:

```python

    pass

```

Output:

```text
10
Parent method
```

The `Child` object can access `x` and `show()` because they were inherited from `Parent`.

# Method Overriding

A child class can also **replace/redefine** a method inherited from the parent.

This is called **method overriding**.

```python
class Vehicle:

    def start(self):
        print("Vehicle started")


class Car(Vehicle):

    def start(self):
        print("Car starts with a key")


car1 = Car()

car1.start()
```

Output:

```text
Car starts with a key
```

Even though `Vehicle` has `start()`, Python uses the `Car` version because the child has overridden it.

---

# Why Does Python Use the Child's Method?

When we do:

```python
car1.start()
```

Python first looks inside the `Car` class.

It finds:

```python
def start(self):
```

So it uses that method.

If `Car` didn't have `start()`, Python would look in the parent:

```text
car1.start()
    ↓
Look in Car
    ↓
Not found?
    ↓
Look in Vehicle
    ↓
Found!
    ↓
Execute Vehicle.start()
```

````python
class Vehicle:
    
    def start(self):
        print("Start the Car")
        
class Car:
    
    def start(self):
        print("Start the Car with a key")


car1 = Car()
car1.start()
````

# `super()` in Inheritance

Sometimes we want to use the parent's method from the child.

Python provides:

```python
super()
```

Example:

```python
class Vehicle:

    def start(self):
        print("Vehicle started")


class Car(Vehicle):

    def start(self):
        super().start()
        print("Car starts moving")


car1 = Car()

car1.start()
```

Output:

```text
Vehicle started
Car starts moving
```

Here:

```python
super().start()
```

means:

> Call the `start()` method of the parent class.

---

````python
class Vehicle:
    
    def start(self):
        print("Start the Car")
        
class Car(Vehicle):
    
    def start(self):
        super().start()
        print("Car starts moving")


car1 = Car()
car1.start()
````

# Inheritance and Constructor

Suppose the parent has a constructor:

```python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):

    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model
```

Now:

```python
car1 = Car("Tata", "Nexon")
```

The following happens:

```python
super().__init__(brand)
```

calls:

```python
Vehicle.__init__(brand)
```

So:

```text
Car object
│
├── brand = "Tata"     ← initialized by Parent
└── model = "Nexon"    ← initialized by Child
```

---

````python
class Vehicle:
    
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model
        
car = Car("Tata", "Nexon")
vehicle = Vehicle("Suzuki")
print(car.brand)
print(vehicle.brand)

# if u want to access brand from vehicle through car object u need to use aggregation

````

# Types of Inheritance in Python

Python mainly supports **5 types of inheritance**:

1. Single Inheritance
2. Multilevel Inheritance
3. Multiple Inheritance
4. Hierarchical Inheritance
5. Hybrid Inheritance

---

# 1. Single Inheritance

**One parent → One child**

```text
Parent
  ↓
Child
```

### Example

```python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


car1 = Car()

car1.start()
car1.drive()
```

Output:

```text
Vehicle starts
Car is driving
```

Here:

```text
Vehicle
   ↓
  Car
```

`Car` inherits from `Vehicle`.

### Real-world example

```text
Animal
   ↓
 Dog
```

A dog **IS-A** animal.
```


````python
class Vehicle:
    
    def start(self):
        print("Vehicle starts")
        
class Car(Vehicle):
    
    def drive(self):
        print("Car starts moving")

car = Car()

car.start()
car.drive()
````

# 2. Multilevel Inheritance

**Grandparent → Parent → Child**

```text
A
↓
B
↓
C
```

### Example

```python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


class SportsCar(Car):

```

Output:

```text
Vehicle starts
Car is driving
Sports car is racing
```

Here:

```text
Vehicle
   ↓
 Car
   ↓
SportsCar
```

`SportsCar` gets:

* `start()` from `Vehicle`
* `drive()` from `Car`
* `race()` from itself

### Easy way to remember

```text
Grandparent
     ↓
  Parent
     ↓
   Child
```

````python
class Vehicle:
    def start(self):
        print("Vehicle starts")
        
class Car(Vehicle):
    def drive(self):
        print("Car starts moving")
        
class SportsCar(Car):
    def race(self):
        print("Car starts racing")
        
sportsCar = SportsCar()

sportsCar.start()
sportsCar.drive()
sportsCar.race()
````

# 3. Multiple Inheritance

**Multiple parents → One child**

```text
Parent1     Parent2
    \         /
     \       /
       Child
```

### Example

```python
class Camera:

    def take_photo(self):
        print("Taking photo")


class Phone:

    def make_call(self):
        print("Making call")


class Smartphone(Camera, Phone):

phone1.make_call()
phone1.browse_internet()
```

Output:

```text
Taking photo
Making call
Browsing internet
```

Here:

```text
Camera       Phone
   \          /
    \        /
    Smartphone
```

`Smartphone` inherits from **both** `Camera` and `Phone`.

So it can use methods from both:

```python
phone1.take_photo()
phone1.make_call()
```

````python
class Camera:
    
    def take_photo(self):
        print("Taking Photo")
        
class Phone:
    
    def make_call(self):
        print("Making Call")
        
class SmartPhone(Camera, Phone):
    
    def browse_internet(self):
        print("Browsing Internet")
        
smartPhone = SmartPhone()

smartPhone.take_photo()
smartPhone.make_call()
smartPhone.browse_internet()
````

# 4. Hierarchical Inheritance

**One parent → Multiple children**

```text
        Parent
        /    \
       ↓      ↓
    Child1  Child2
```

### Example

```python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


class Bike(Vehicle):

    def ride(self):
        print("Bike is riding")


car1 = Car()
bike1 = Bike()

car1.start()
car1.drive()

bike1.start()
bike1.ride()
```

Output:

```text
Vehicle starts
Car is driving
Vehicle starts
Bike is riding
```

Here:

```text
          Vehicle
          /     \
         ↓       ↓
       Car      Bike
```

Both `Car` and `Bike` inherit from `Vehicle`.

Therefore both can use:

```python
start()
```

But:

```text
Car → drive()
Bike → ride()
```

are their own methods.

---

````python
class Vehicle:
    def start(self):
        print("Vehicle starts")
        
class Car(Vehicle):
    def drive(self):
        print("Car starts moving")
        
class Bike(Vehicle):
    def ride(self):
        print("Bike is riding")
        
car = Car()
bike = Bike()

car.start()
car.drive()

bike.start()
bike.ride()
````

# 5. Hybrid Inheritance

**Combination of two or more types of inheritance.**

For example, we can combine **multiple + multilevel inheritance**.

```text
        Vehicle
        /     \
       ↓       ↓
     Car      Bike
       \       /
        \     /
        Racing
```

### Example

```python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


class Bike(Vehicle):

    def ride(self):
        print("Bike is riding")


class Racing(Car, Bike):

    def race(self):
        print("Racing")


r1 = Racing()

r1.start()
r1.drive()
r1.ride()
r1.race()
```

Output:

```text
Vehicle starts
Car is driving
Bike is riding
Racing
```

Here we have:

```text
             Vehicle
             /     \
            /       \
          Car       Bike
            \       /
             \     /
             Racing
```

This combines:

* **Hierarchical inheritance** → `Vehicle → Car` and `Vehicle → Bike`
* **Multiple inheritance** → `Racing(Car, Bike)`

Therefore it is **Hybrid Inheritance**.

````python
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


class Bike(Vehicle):

    def ride(self):
        print("Bike is riding")


class Racing(Car, Bike):

    def race(self):
        print("Racing")


r1 = Racing()

r1.start()
r1.drive()
r1.ride()
r1.race()
````

# Inheritance with Private Variables

Private variables are variables whose names start with **double underscore `__`**.

Example:

```python
class Parent:

    def __init__(self):
        self.__balance = 5000
```

Here:

```text
__balance
    ↓
Private variable
```

But there is an important point when **inheritance** is involved.

---

# Example

```python
class Parent:

    def __init__(self):
        self.__balance = 5000

    def show_balance(self):
        print(self.__balance)


class Child(Parent):
    pass


child1 = Child()

child1.show_balance()
```

Output:

```text
5000
```

You might ask:

> If `__balance` is private in `Parent`, how can `Child` access it?

The answer is:

**The child cannot directly access the parent's private variable.**

But the child **can use a parent method that accesses it**.

Here:

```python
def show_balance(self):
    print(self.__balance)
```

is a method of the parent.

So:

```python
child1.show_balance()
```

calls the parent's method, and that method can access its own private variable.

---

# Direct Access from Child

Now try this:

```python
class Parent:

    def __init__(self):
        self.__balance = 5000


class Child(Parent):

    def show(self):
        print(self.__balance)


child1 = Child()

child1.show()
```

This gives:

```text
AttributeError
```

Why?

Because:

```python
self.__balance
```

inside `Parent` is name-mangled to:

```python
self._Parent__balance
```

But inside `Child`, Python interprets:

```python
self.__balance
```

as:

```python
self._Child__balance
```

These are **different names**.

```text
Parent:

__balance
    ↓
_Parent__balance


Child:

__balance
    ↓
_Child__balance
```

So the child cannot directly access the parent's private variable using `self.__balance`.

---

# But Protected Variables Are Different

Python also has a convention using a **single underscore**:

```python
class Parent:

    def __init__(self):
        self._balance = 5000


class Child(Parent):

    def show(self):
        print(self._balance)


child1 = Child()

child1.show()
```

Output:

```text
5000
```

Here `_balance` can be accessed by the child.

So:

```text
__balance  → Private/name-mangled
_balance   → Protected convention
balance    → Public
```

# Important Examples

## Case 1 — `pqr` has no `__init__()`

```python
class abc:
    def __init__(self, value):
        self.__value = value

    def get_value(self):
        return self.__value


class pqr(abc):
    pass


a = pqr(20)

print(a.get_value())
```

This works.

Why?

Because `pqr` **doesn't have its own `__init__()`**.

So Python uses the parent's:

```python
abc.__init__()
```

Therefore:

```python
a = pqr(20)
```

causes the parent constructor to run:

```python
self.__value = 20
```

So the object has:

```text
a
└── _abc__value = 20
```

Then:

```python
a.get_value()
```

works because `get_value()` finds the parent's private variable.


````python
# Example 1:

class abc:
    def __init__(self, value):
        self.__value = value
        
    def get_value(self):
        return self.__value

class pqr(abc):

    def greet(self):
        print("Hello")
        
a = pqr(10)

a.get_value()
````

# Case 2 — `pqr` has its own `__init__()`

Your second code:

```python
class abc:
    def __init__(self, value):
        self.__value = value

    def get_value(self):
        return self.__value


class pqr(abc):

    def __init__(self, num, value):
        self.num = num

    def get_num(self):
        return self.__num


a = pqr(10, 20)

a.get_value()
```

Now `pqr` has its own:

```python
def __init__(self, num, value):
    self.num = num
```

So when you do:

```python
a = pqr(10, 20)
```

Python calls:

```python
pqr.__init__(a, 10, 20)
```

It does **not automatically call**:

```python
abc.__init__()
```

Therefore this line from the parent:

```python
self.__value = value
```

never executes.

So `a` does **not** have the parent's `__value`.

Then you call:

```python
a.get_value()
```

The inherited `get_value()` runs:

```python
return self.__value
```

But `__value` was never created.

So you get an `AttributeError`.


````python
# Example 2:

class abc:
    def __init__(self, value):
        self.__value = value
        
    def get_value(self):
        return self.__value

class pqr(abc):
    
    def __init__(self, num, value):
        self.num = num
        
    def get_num(self):
        return self.__num
        
a = pqr(10, 20)

# a.get_value() -> this will give error
````

## If You Want Both Parent and Child Data

You need to make the parent's constructor run as well:

```python
class abc:
    def __init__(self, value):
        self.__value = value

    def get_value(self):
        return self.__value


class pqr(abc):

    def __init__(self, num, value):
        super().__init__(value)
        self.num = num


a = pqr(10, 20)

print(a.get_value())
print(a.num)
```

Output:

```text
20
10
```

So the key rule to remember is:

> **If a child class doesn't define `__init__()`, it can use the parent's `__init__()`. If the child defines its own `__init__()`, the parent's `__init__()` does not run automatically.**


````python
# Example 3
class abc:
    def __init__(self, value):
        self.__value = value
        
    def get_value(self):
        return self.__value

class pqr(abc):
    
    def __init__(self, num, value):
        super().__init__(value)
        self.__num = num
        
    def get_num(self):
        return self.__num
        
a = pqr(10, 20)

print(a.get_value())
print(a.get_num())
````
