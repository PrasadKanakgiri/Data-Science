```python
import random

number = random.randint(1,100)
# print(number)

x = int(input("Guess the number: "))
attempts = 1
   
while x != number:
    if x > number:
        print("Too Big!!!, Enter small number")   
    else:
        print("Too Small!!!, Enter big number")
        
    attempts+=1
    x = int(input("Guess the number: "))

print("You guessed it right!!!, you took", attempts, "attempts")

        
```
