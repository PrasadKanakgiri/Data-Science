# Lambda Functions in Python

A **lambda function** is a small, anonymous function written in a single line.

## 1. Normal Function

```python
def square(x):
    return x * x

print(square(5))
```
#### Output:

25

## The same function using a lambda:
```python
square = lambda x: x * x

print(square(5))
```
#### Output:

25

# Lambda Syntax

```python
lambda arguments: expression
```
Example:
```python
lambda x: x * 2
```
##### Here:

lambda       → tells Python we are creating a lambda function


x            → argument

:            → separates argument from expression

x * 2        → expression

It is roughly equivalent to:
```python
def something(x):
    return x * 2
```


````python
sqaures = lambda x : x*x
print(sqaures(10))
````

# 3. Multiple Arguments

A `lambda` can accept multiple arguments:
```python
add = lambda a, b: a + b

print(add(10, 20))
```
Output:

30

##### Equivalent normal function:
```python
def add(a, b):
    return a + b
```

````python
addition = lambda a,b : a + b
print(addition(3,6))
````

# 4. Lambda with a Condition
```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(5))
print(check(8))
```
#### Output:

Odd
Even

##### Equivalent normal function:
```python
def check(x):
    if x % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

````python
check = lambda x : "Even" if x%2 == 0 else "Odd"
print(check(10))
print(check(5))
````

# 5. Important Point

A `lambda` can contain only one expression.

The result of that expression is automatically returned.
```python
square = lambda x: x * x
```
You don't need to write:
```python
return x * x
```
because `lambda` automatically returns the expression's result.

##### Remember

#### Normal function:
```python
def square(x):
    return x * x

```
#### Lambda:
```python
lambda x: x * x
```
**So, lambda = a short way of creating a small function.**
