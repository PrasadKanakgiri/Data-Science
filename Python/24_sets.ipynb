# Set

A `set` is an **unordered and mutable collection of unique values**.

Sets are created using **curly braces `{}`**.

A set **does not allow duplicate values**.

### Important Points
- Sets are unordered.
- Sets are mutable.
- Sets contain unique values.
- Sets do not support indexing.
- Sets do not support slicing.
- Sets do not allow mutable datatypes like list.
- Set elements must be hashable.
- Multi dimensional sets are not possible as sets itself are mutable and they donot allow mutable datatypes

### Creating a Set

```python
numbers = {1, 2, 3, 4, 5}

print(numbers)
```
### Output:

{1, 2, 3, 4, 5}

**If duplicate values are added, they are automatically removed.**
```python
numbers = {1, 2, 2, 3, 3, 4}

print(numbers)
```
### Output:

{1, 2, 3, 4}


````python
num = {1, 2 , 3, 4}
print(num)
print(type(num))

num2 = {1, 2 , 3, 3, 4, 4, 5, 5}
print(num2)    # remove the duplicate elements
````

### Empty Set

To create an **empty set**, use `set()`.
```python
numbers = set()

print(numbers)
print(type(numbers))
```
### Output:

set()
<class 'set'>

**`{}` creates an empty dictionary, not an empty set.**

````python
s1 = {}
print(type(s1))  # this is a dictionary

s2 = set()
print(type(s2))  # this is a set
````

### Adding Elements

The `add()` method is used to **add a single element to a set**.
```
numbers = {1, 2, 3}

numbers.add(4)

print(numbers)
```
### Output:

{1, 2, 3, 4}


````python
num = {1, 2, 3}
print(num)
num.add(4)
print(num)
````

### Updating a Set

The `update()` method is used to **add multiple elements to a set**.
```python
numbers = {1, 2, 3}

numbers.update([4, 5, 6])

print(numbers)
```
### Output:

{1, 2, 3, 4, 5, 6}

````python
num = {1, 2, 3}
print(num)
num.update([4, 5, 6])
print(num)
````

### Removing Elements

The `remove()` method **removes a specific value from a set**.
```python
numbers = {1, 2, 3, 4}

numbers.remove(3)

print(numbers)
```
### Output:

{1, 2, 4}

**If the value does not exist, `remove()` raises a KeyError.**

### discard()

The `discard()` method also **removes a specific value**.

**Unlike `remove()`, it does not raise an error if the value does not exist.**
```python
numbers = {1, 2, 3, 4}

numbers.discard(5)

print(numbers)
```
### Output:

{1, 2, 3, 4}

### pop()

The `pop()` method **removes and returns an arbitrary element from a set**.
```python
numbers = {10, 20, 30, 40}

x = numbers.pop()

print(x)
print(numbers)
```
**The element removed is not guaranteed to be the first or last element.**

### clear()

The `clear()` method **removes all elements from a set**.
```python
numbers = {1, 2, 3, 4}

numbers.clear()

print(numbers)
```
### Output:

set()

````python
num = {1, 2, 3, 4, 5, 6}
print(num)

num.remove(2)
print(num)
# num.remove(7) -> will throw error as 7 is not present in the set

num.discard(4)
print(num)
num.discard(8) # wont throw error even after 7 not present in the set
print(num)

num.pop()
print(num)

num.clear()
print(num)

````

### Membership

The `in` operator **checks whether an element exists in a set**.
```python
numbers = {10, 20, 30, 40}

print(20 in numbers)
print(50 in numbers)
```
### Output:

True
False

````python
num = {1, 2, 3, 4}

print(2 in num)
print(5 in num)
print(6 not in num)
````

### Set Operations

`Sets` support mathematical operations such as union, intersection, difference, and symmetric difference.
```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print(set1 | set2)
print(set1 & set2)
print(set1 - set2)
print(set1 ^ set2)
```
### Output:

{1, 2, 3, 4, 5, 6}
{3, 4}
{1, 2}
{1, 2, 5, 6}

````python
num1 = {1, 2, 3, 4}
num2 = {3, 4, 6, 5}

print(num1 | num2)
print(num1 & num2)
print(num1 - num2)  # non common elements present in num1 but not num2
print(num2 - num1)  # non common elements present in num2 but not num1
print(num1 ^ num2)  # non common elements present in both num1 and num2
print("-----------------")
print(num1.union(num2))
print(num1.intersection(num2))
print(num1.difference(num2))
print(num1.symmetric_difference(num2))
print(num1.isdisjoint(num2))  # no common elemets in either of sets
print(num1.issubset(num2))
print(num1.issuperset(num2))
````
