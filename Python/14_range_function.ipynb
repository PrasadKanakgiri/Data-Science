# range() Function

The `range()` function is used to **generate a sequence of numbers**.

It is commonly used with `for` loops.

### Syntax

```python
range(start, stop, step)
```

`start` → The number from which the sequence `starts`. Default is `0`.

`stop` → The number at which the sequence `stops`. The stop value is not included.

`step` → The amount by which the number increases or decreases. Default is `1`.

````python
range(1,10)
````

````python
list(range(1,10))
````

````python
list(range(10))
````

````python
list(range(1,10,3))
````

````python
list(range(10,0))
````

````python
list(range(10,0,-1))
````

````python
list(range(10,0,-3))
````
