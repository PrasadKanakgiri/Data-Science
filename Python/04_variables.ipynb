# Variables

A variable is a **name used to store a value** in Python.

A variable is created when a value is assigned to it using the `=` operator.

```python
name = "John"
age = 10
height = 5.8
is_student = True

````python
name = "John"
print(name)
````

### Dynamic Typing

Python is a **dynamically typed** language.

This means you do not need to specify the data type of a variable when creating it. Python automatically determines the type based on the value assigned.

```python
x = 10
print(x)

x = "India"
print(x)

### Static Typing

In a **statically typed** language, the data type of a variable is specified when the variable is declared.

The variable is generally expected to store values of the specified type.

For example, in Java:

```java
int x = 10;

### Dynamic Binding

In Python, **dynamic binding** means that a variable is bound to an object at **runtime**.

The same variable can refer to objects of different data types during the execution of the program.

```python
x = 10
print(x)

x = "India"
print(x)

x = 3.5
print(x)

````python
x = 10
print(x)
````

````python
x = "India"
print(x)
````

````python
x = 3.5
print(x)
````

````python
print(x)
````

### Static Binding

Static binding means that a variable is **fixed to a particular type** when it is declared.

The variable is expected to hold values of the **same type**.

For example, in Java:

```java
int x = 10;

Here, x is an integer and is expected to store integer values.

### Assigning Multiple Variables

Python allows us to assign values to **multiple variables** in different ways.


### Assigning Values Separately

We can assign values to each variable separately.

```python
a = 5
b = 6
c = 7

print(a)
print(b)
print(c)

````python
a = 5; b = 6; c = 7
print(a)
print(b)
print(c)
````

### Multiple Assignment

Python allows us to assign **multiple values to multiple variables in a single line**.

```python
a, b, c = 5, 6, 7

Here, 5 is assigned to a, 6 to b, and 7 to c.

print(a)
print(b)
print(c)

````python
a, b, c = 5, 6, 7
print(a)
print(b)
print(c)
````

### Assigning the Same Value

Python allows us to assign the **same value to multiple variables in a single line**.

```python
a = b = c = 5

Here, 5 is assigned to a, b, and c.

print(a)
print(b)
print(c)

````python
a = b = c = 5
print(a)
print(b)
print(c)
````
