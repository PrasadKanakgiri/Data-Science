# List

A `list` is an **ordered and mutable collection** of values.

A list can store **multiple values of different data types**.

Lists are created using **square brackets `[]`**.

### Arrays vs List
1. `Arrays` are homogeneous i.e., it can contain values of same datatype only while `Lists` can be hetrogeneous.
2. `Arrays` store the values in continuos order at memory level.
3. Due to its continuos memory allocation, `Arrays` are faster than `Lists`.
4. `Lists` are more programmer friendly.

### Creating a List

```python
numbers = [1, 2, 3, 4, 5]

print(numbers)
```
### Output:

[1, 2, 3, 4, 5]

A `list` can contain values of different data types.
```python
data = [10, 3.5, True, "Python"]

print(data)
```
### Output:

[10, 3.5, True, 'Python']
