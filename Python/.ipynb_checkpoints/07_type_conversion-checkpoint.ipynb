# Type Conversion

Type conversion means **converting a value from one data type to another**.

There are two types of type conversion in Python:

1. **Implicit Type Conversion**
2. **Explicit Type Conversion**

### 1. Implicit Type Conversion

Implicit type conversion is when Python **automatically converts** one data type into another.

```python
a = 5
b = 2.5

c = a + b

print(c)
print(type(c))

````python
a = 5
b = 2.5
c = a  + b
````

````python
print(c)
````

Here, Python automatically converts the integer 5 into a float before performing the calculation.

````python
print(type(c))
````

### Explicit Type Conversion

Explicit type conversion is when we **manually convert** one data type into another using functions such as `int()`, `float()`, `str()`, and `bool()`.

```python
a = "10"

b = int(a)

print(b)
print(type(b))

````python
a = "10"
b = int(a)
````

````python
print(b)
````

````python
print(type(b))
````
