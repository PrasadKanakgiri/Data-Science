# for Loop

A `for` loop is used to **repeat a block of code for each item in a sequence or collection**.

### Syntax

```python
for variable in sequence:
    # code to execute
```

### Example
```python
for i in range(5):
    print(i)
```
### Output:

0
1
2
3
4

Here, the for loop takes each value from range(5) one by one and stores it in the variable i.

#### Example with a List
```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

### Output:

Apple
Banana
Mango

Here, the loop goes through each item in the list one by one.

````python
for i in range(1,10):
    print(i)
````

````python
for i in "India":
    print(i)
````

````python
for i in range(1,10):
    print(i, sep="-")
````

````python
for i in range(1,10):
    print(i, end="---")
````

# `sep` vs `end` in Python

## Why didn't `sep` work in a `for` loop?

Consider the following code:

```python
for i in range(1, 10):
    print(i, sep="---")
```

Output:

```text
1
2
3
4
5
6
7
8
9
```

### Why?

`sep` is used to **separate multiple values passed to a single `print()` statement**, not **multiple calls to a `print()`. statement**

In our code:

```python
print(i, sep="---")
```

`print()` receives only **one value**: `i`.

Therefore, there is nothing for `sep` to separate.

---

# `sep` Parameter

`sep` stands for **separator**.

It is used to specify what should be printed **between multiple values** in a single `print()` statement.

### Example

```python
print(1, 2, 3, sep="---")
```

Output:

```text
1---2---3
```

Here, `print()` receives three values:

```text
1    2    3
```

So `sep="---"` is placed between them:

```text
1 --- 2 --- 3
```

### Default value of `sep`

By default:

```python
sep = " "
```

For example:

```python
print(1, 2, 3)
```

Output:

```text
1 2 3
```

---

# `end` Parameter

If you want to add something **after every `print()` call**, use the `end` parameter.

For example:

```python
for i in range(1, 10):
    print(i, end="---")
```

Output:

```text
1---2---3---4---5---6---7---8---9---
```

### Why does `end` work?

Every iteration calls `print()` separately:

```text
print(1)
print(2)
print(3)
...
```

The `end` parameter specifies what should be printed **at the end of each `print()` call**.

By default:

```python
end = "\n"
```

So normally:

```python
print(1)
print(2)
```

produces:

```text
1
2
```

because `print()` adds a new line after each call.

When we use:

```python
print(1, end="---")
print(2, end="---")
```

we get:

```text
1---2---
```

---

# `sep` vs `end`

| Parameter | Purpose                                            | Example                     |
| --------- | -------------------------------------------------- | --------------------------- |
| `sep`     | Separates multiple values **inside one `print()`** | `print(1, 2, 3, sep="---")` |
| `end`     | Specifies what comes **after each `print()` call** | `print(1, end="---")`       |

### Easy way to remember

```text
sep → between values
end → after print
```

### Example using both

```python
print(1, 2, 3, sep="---", end="!!!")
```

Output:

```text
1---2---3!!!
```

Here:

* `sep="---"` → puts `---` **between** `1`, `2`, and `3`
* `end="!!!"` → puts `!!!` **after** the entire `print()` statement
