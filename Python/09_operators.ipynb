### Operators

Python has different types of operators:

1. **Arithmetic Operators** → Used to perform mathematical operations.
2. **Relational (Comparison) Operators** → Used to compare two values.
3. **Logical Operators** → Used to combine or reverse conditions.
4. **Assignment Operators** → Used to assign values to variables.
5. **Membership Operators** → Used to check whether a value exists in a sequence.
6. **Identity Operators** → Used to check whether two variables refer to the same object.
7. **Bitwise Operators** → Used to perform operations on individual bits.



# 1. Arithmetic Operators

Arithmetic operators are used to **perform mathematical operations** on values.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `5 + 2` | `7` |
| `-` | Subtraction | `5 - 2` | `3` |
| `*` | Multiplication | `5 * 2` | `10` |
| `/` | Division | `5 / 2` | `2.5` |
| `//` | Floor Division | `5 // 2` | `2` |
| `%` | Modulus | `5 % 2` | `1` |
| `**` | Exponentiation | `5 ** 2` | `25` |

### Example

```python
a = 5
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)  # same as division, just converted the result into int
print(a % b)
print(a ** b)

````python
a = 5
b = 2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
````

# 2. Comparison Operators

Comparison operators are used to **compare two values**.

They return either `True` or `False`.

| Operator | Name | Example | Result |

| `!=` | Not equal to | `5 != 2` | `True` |
| `>` | Greater than | `5 > 2` | `True` |
| `<` | Less than | `5 < 2` | `False` |
| `>=` | Greater than or equal to | `5 >= 2` | `True` |
| `<=` | Less than or equal to | `5 <= 2` | `False` |

### Example

```python
a = 5
b = 2

print(a != b)

# 3. Logical Operators

Logical operators are used to **combine or modify conditions**.

They return either `True` or `False`.

| Operator | Name | Example | Result |

```python

# 4. Bitwise Operators

Bitwise operators are used to **perform operations on the individual bits of numbers**.

| Operator | Name | Example | Result |
|---|---|---|---|
| `&` | Bitwise AND | `5 & 3` | `1` |
| `|` | Bitwise OR | `5 | 3` | `7` |
| `^` | Bitwise XOR | `5 ^ 3` | `6` |
| `~` | Bitwise NOT | `~5` | `-6` |
| `<<` | Left Shift | `5 << 1` | `10` |
| `>>` | Right Shift | `5 >> 1` | `2` |

### Example

```python
a = 5
b = 3

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 1)
print(a >> 1)

````python
a = 2
b = 3

print(a & b)
# 010 -> 2
# 110 -> 3
# --- -> &
# 010 -> 2
print(a | b)
# 010 -> 2
# 110 -> 3
# --- -> |
# 110 -> 3
print(a ^ b) 
# 010 -> 2
# 110 -> 3
# --- -> ^
# 100 -> 4  same bit give 0 and different bit give 1
print(~a)
# 010 -> 2
# --- -> ~
# 101 -> -3  ~n = -(n+1)  so ~2 = -(2+1) = -3 
print(a << 1)
# 010 -> 2
# --- -> <<
# 100 -> 4
print(a >> 1)
# 010 -> 2
# --- -> >>
# 001 -> 1
````

Assignment operators are used to **assign values to variables**.

| Operator | Name | Example | Equivalent To |

# Identity Operators

Identity operators are used to **check whether two variables refer to the same object in memory**.

They return either `True` or `False`.

| Operator | Name | Example | Result |
|---|---|---|---|
| `is` | Identity | `a is b` | `True` if both refer to the same object |
| `is not` | Not Identity | `a is not b` | `True` if both refer to different objects |

### Example

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a is not c)

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a is not c)
````

# Membership Operators

Membership operators are used to **check whether a value exists in a sequence or collection**.

They return either `True` or `False`.

| Operator | Name | Example | Result |
|---|---|---|---|
| `in` | Membership | `2 in [1, 2, 3]` | `True` |
| `not in` | Not Membership | `5 not in [1, 2, 3]` | `True` |

### Example

```python
a = [1, 2, 3, 4, 5]

print(3 in a)
print(6 in a)
print(3 not in a)
print(6 not in a)

````python
a = [1, 2, 3, 4, 5]

print(3 in a)
print(6 in a)
print(10 in a)
print(3 not in a)
````
