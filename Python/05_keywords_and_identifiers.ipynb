# Keywords

Keywords are **reserved words in Python** that have a special meaning.

They cannot be used as variable names, function names, or other identifiers.

Some common Python keywords are:

```python
if
else
for
while
def
class
return
True
False
None
and
or
not

# Identifiers

Identifiers are the **names given to variables, functions, classes, and other objects** in Python.

Examples:

```python
name = "John"
age = 10


### Rules for Naming Identifiers

- An identifier can contain **letters, numbers, and underscores (`_`)**.
- An identifier **cannot start with a number**.
- An identifier **cannot contain spaces**.
- An identifier **cannot be a Python keyword**.
- Identifiers are **case-sensitive**.

```python
first_name = "John"   # Valid
age2 = 10             # Valid
_first_name = "John"  # Valid

2age = 10             # Invalid
first name = "John"   # Invalid
if = 10               # Invalid
first-name = "John"   # Invalid
