# Polymorphism

**Polymorphism** means:

> **One thing can have many forms.**

In Python, polymorphism allows the **same method name or operation to behave differently depending on the object**.

---

## Simple Example

```python
class Dog:
    def sound(self):
        print("Dog barks")


class Cat:
    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

**Output:**

```text
Dog barks
Cat meows
```

Both classes have the **same method name**:

```python
sound()
```

But the behavior is different.

* `dog.sound()` → Dog barks
* `cat.sound()` → Cat meows

This is **polymorphism**.

---

## Polymorphism with Inheritance

A common example is **method overriding**.

```python
class Vehicle:
    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    def start(self):
        print("Car starts with a key")


class Bike(Vehicle):
    def start(self):
        print("Bike starts with a button")


car = Car()
bike = Bike()

car.start()
bike.start()
```

**Output:**

```text
Car starts with a key
Bike starts with a button
```

The same method:

```python
start()
```

behaves differently for different objects.

---

## The Main Idea

Think of it like this:

```text
             start()
                |
        -----------------
        |               |
       Car             Bike
        |               |
   Key start       Button start
```

The **method name is the same**, but the **behavior changes**.

---
## Real-Life Example

Imagine a **person pressing a "Start" button**.

* In a car → engine starts
* In a bike → engine starts
* In a washing machine → washing starts

The action is similar:

```text
Start
```

But the behavior is different depending on the object.

That's the basic idea of **polymorphism**.

---

## Four Common Forms of Polymorphism in Python

You may encounter:

1. **Method overriding** → child provides its own version of parent's method
2. **Duck typing** → object is used based on what it can do, not its class
3. **Operator overloading** → `+`, `-`, `==`, etc. behave differently for different types
4. **Method overloading** → Python doesn't support traditional method overloading directly; it is usually achieved using default arguments or `*args`

```text
MORPH = Forms

```

# Duck Typing

**Duck typing** is a concept in Python where we care about **what an object can do**, rather than **what type/class the object belongs to**.

The idea comes from the saying:

> **"If it walks like a duck and quacks like a duck, then it is a duck."**

In Python:

> **If an object has the required method, we can use it.**

---

## Simple Example

```python
class Dog:
    def sound(self):
        print("Dog barks")


class Cat:
    def sound(self):
        print("Cat meows")


def make_sound(animal):
    animal.sound()


dog = Dog()
cat = Cat()

make_sound(dog)
make_sound(cat)
```

**Output:**

```text
Dog barks
Cat meows
```

Notice something important:

```python
def make_sound(animal):
    animal.sound()
```

We don't check:

```python
if type(animal) == Dog:
```

or:

```python
if type(animal) == Cat:
```

We simply assume that `animal` has a `sound()` method.

If it does → it works.

---

## Another Example

```python
class Car:
    def start(self):
        print("Car starts")


class Bike:
    def start(self):
        print("Bike starts")


def start_vehicle(vehicle):
    vehicle.start()


car = Car()
bike = Bike()

start_vehicle(car)
start_vehicle(bike)
```

**Output:**

```text
Car starts
Bike starts
```

The function doesn't care whether `vehicle` is a `Car` or `Bike`.

It only cares:

> **"Does this object have a `start()` method?"**

---

## What If the Method Doesn't Exist?

```python
class Bicycle:
```

# Duck Typing vs Inheritance

### Inheritance

With inheritance, we have a relationship:

### Duck Typing

With duck typing, there doesn't need to be any inheritance relationship:

```text

```

Both objects simply provide the method we need.

# Operator Overloading

**Operator overloading** means giving an operator such as `+`, `-`, `*`, `==`, etc. a specific behavior for objects of a class.

For example:

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


num1 = Number(10)
num2 = Number(20)

print(num1 + num2)
```

**Output:**

```text
30
```

Here:

```python
num1 + num2
```

internally uses:

```python
num1.__add__(num2)
```

We defined what `+` should do when used with our `Number` objects.

### Another Example

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks


s1 = Student(80)
s2 = Student(90)

print(s1 + s2)
```

**Output:**

```text
170
```

So:

```text
+ operator
     ↓
__add__()
     ↓
defines what + means for our objects
```

### Common Operator Methods

| Operator | Special method  |
| -------- | --------------- |
| `+`      | `__add__()`     |
| `-`      | `__sub__()`     |
| `*`      | `__mul__()`     |
| `/`      | `__truediv__()` |
| `==`     | `__eq__()`      |
| `<`      | `__lt__()`      |
| `>`      | `__gt__()`      |

### Easy Memory Trick

> **Operator overloading = changing the behavior of an operator for objects.**

````python
class Student:
    
    def __init__(self, marks):
        self.marks = marks
        
    def __add__(self, other):
        return self.marks + other.marks

s1 = Student(50)
s2 = Student(30)

print(s1 + s2)
````

# Method Overloading

**Method overloading** means having the **same method name** but allowing it to work with **different numbers or types of arguments**.

For example, in some languages you can write:

```text
add(a, b)
add(a, b, c)
```

Both have the same method name `add()`, but accept different numbers of arguments.

### Does Python support traditional method overloading?

**No.**

Python does not support traditional method overloading like Java or C++.

If you define the same method twice:

```python
class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c
```

The second `add()` **replaces** the first one.

So this:

```python
calc = Calculator()

print(calc.add(10, 20))
```

will give an error because the final `add()` expects 3 arguments.

---

# How do we achieve similar behavior in Python?

We commonly use **default arguments**.

```python
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

**Output:**

```text
30
60
```

The same method:

```python
add()
```

works with either 2 or 3 arguments.

Another way is using `*args`:

```python
class Calculator:

    def add(self, *args):
        return sum(args)


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
```

**Output:**

```text
30
60
100
```


````python
class Calculator:
    
    def add(self, *args):
        return sum(args)

calc = Calculator()

print(calc.add(1, 2))
print(calc.add(1, 2, 3))
print(calc.add(1, 2, 3, 4))
print(calc.add(1, 2, 3, 4, 5))
````


# Operator Overloading vs Method Overloading

|                     | Operator Overloading | Method Overloading          |
| ------------------- | -------------------- | --------------------------- |
| What is overloaded? | Operator             | Method                      |
| Example             | `+`                  | `add()`                     |
| Python support      | ✅ Yes                | ❌ No traditional support    |
| Common technique    | Dunder methods       | Default arguments / `*args` |
| Example             | `__add__()`          | `add(a, b, c=0)`            |

### Remember

```text
Operator Overloading
        ↓
Same operator → different behavior

Method Overloading
        ↓
Same method → different arguments
```

For Python specifically:

> **Operator overloading is directly supported through special methods such as `__add__()`. Traditional method overloading is not directly supported; default arguments and `*args` are commonly used instead.**
