# Modules

A module is a **file containing Python code, functions, and classes** that can be used in another Python program.

Python provides many **built-in modules** that can be imported and used directly.

### Common Python Modules

| Module | Purpose |
|---|---|
| `math` | Provides mathematical functions |
| `random` | Used to generate random numbers and make random choices |
| `os` | Used to interact with the operating system |
| `time` | Used to work with time-related functions |

### Importing a Module

A module can be imported using the `import` keyword.

```python
import math
import random
import os
import time

````python
help('modules')
````

````python
import math

print(math.sqrt(25))
print(math.pi)
print(math.e)
print(math.factorial(5))
print(math.ceil(4.1))
print(math.floor(4.9))
````

````python
import random

print(random.randint(1,100))

a = [1,2,3,4,5]
random.shuffle(a)
a
````

````python
import time

print(time.time())

print(time.ctime())  # current time

print("Hello")
time.sleep(3)
print("World")
````

````python
import os

print(os.getcwd())
(os.listdir())
````
