# print() Function

The `print()` function is used to **display output** on the screen.

## 1. Printing Different Data Types

The `print()` function can print different types of values:
```python
print("Hello World")   # String
print(6)              # Integer
print(3.5)            # Float
print(False)          # Boolean

````python
print("Hello World")
print(6)
print(3.5)
print(False)
````

# Printing Multiple Values

The `print()` function can print multiple values by separating them with commas.
```python
print("India", "Sri Lanka", "Pakistan")

````python
print("India", "Sri Lanka", "Pakistan")
````

````python
print("India", 5, True)
````

# sep Parameter

The `sep` parameter is used to specify the separator between multiple values.

By default, `sep`  is a space " ".
```python
print("India", "Sri Lanka", "Pakistan", sep="---")

````python
print("India", "SriLanka", sep="---")
````

### `end`

The `end` parameter is used to specify what should be printed at the end of the output.

By default, `end` is a new line `"\n"`.
```python
print("Hello")
print("World")

````python
print("Hello")
print("World")
````

````python
print("Hello", end=" ")
print("world")
````
