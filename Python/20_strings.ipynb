# Strings

A string is a **sequence of characters**.

In Python specifically, strings are **sequence of unicode characters**.

Strings are written using **single quotes (`' '`), double quotes (`" "`), or triple quotes (`''' '''` / `""" """`)**.

```python
name = "Python"
city = 'Pune'
message = """Hello World"""

### Adding Characters to a String

Strings are **immutable**, which means we cannot directly add a character to an existing string.

We can create a new string by using the `+` operator.

```python
name = "Pytho"

name = name + "n"

print(name)
```

### Important

Strings are **immutable**, which means we cannot change an existing string.

When we add a character to a string, Python creates a **new string object** with a different memory address.

```python
name = "Pytho"

print(id(name))

name = name + "n"

print(name)
print(id(name))
```
The two `id()` values will generally be different because a **new string object is created**.

The variable `name` is then made to refer to the new string.

Important: The original string `"Pytho"` is **not modified**.


So conceptually:

```text
name → "Pytho"       (old object)
          ↓
      + "n"
          ↓
name → "Python"      (new object)

````python
name = "Pytho"
print(id(name))

name = name + "n"
print(name)
print(id(name))
````

### Editing a String

Strings are **immutable**, which means we cannot directly change a character in an existing string.

```python
name = "Python"

# name[0] = "J"   # Error
```

Instead, we create a new string with the required changes.
```python
name = "Python"

name = "J" + name[1:]

print(name)
```

### Output:

Jython

Here, the original string `"Python"` is not changed. A new string object `"Jython"` is created.

````python
name1 = "Python"

name2 = "J" + name1[1:]
print(name2)
````

### Deleting a String

Strings are **immutable**, so we cannot delete individual characters from an existing string.

We can delete the **entire string object** using the `del` keyword.

```python
name = "Python"

del name
```
After using `del`, the variable `name` no longer exists.

print(name)

This will give a `NameError` because name has been deleted.

### Deleting a Character

To remove a character, we cannot delete it directly. Instead, we create a new string without that character.

```python
name = "Python"

name = name[:2] + name[3:]

print(name)
```

### Output:

Pyhon

````python
name = "India"
print(name)

del name

# print(name)
````

### String Operations

String operations are used to **perform different operations on strings**.

| Operation | Operator | Example | Result |
|---|---|---|---|
| Concatenation | `+` | `"Hello" + "World"` | `"HelloWorld"` |
| Repetition | `*` | `"Hello" * 3` | `"HelloHelloHello"` |
| Indexing | `[]` | `"Python"[0]` | `"P"` |
| Slicing | `[:]` | `"Python"[0:3]` | `"Pyt"` |
| Membership | `in` | `"P" in "Python"` | `True` |

### Example

```python
a = "Hello"
b = "World"

print(a + b)
print(a * 3)
print(a[0])
print(a[0:3])
print("H" in a)

````python
a = "Hello"
b = " World"

print(a+b)
print(a*3)
print(a[3])
print(a[1:4])
print('e' in a)
````

### Important

Empty string is considered as `False`

Non-Empty string is considered as `True`

"" -> `False`
"abc" -> `True`

### Example
```python

```
### Output
"" # False

```python
"" or "hello" # False or True
```
### Output
"hello" # True

```python
"hello" and "world" # True and True
```
### Output
"world" # it will print 2nd string because the answer will be true only when 2nd string is true

```python
"hello" or "world" # True or True
```
### Output

````python
print("" and "hello")          # ""
print("" or "hello")           # ""
print("hello" or "")           # "hello"
print("hello" or "world")      # "hello"
print("hello" and "world")     # "world"
````

### String Functions

String functions are used to **perform different operations on strings**.

| Function | Purpose | Example |
|---|---|---|
| `len()` | Returns the length of a string | `len("Python")` |
| `max()` | Returns the character with the highest Unicode value | `max("Python")` |
| `min()` | Returns the character with the lowest Unicode value | `min("Python")` |
| `sorted()` | Returns the characters in sorted order | `sorted("Python")` |

### Common String Methods

| Method | Purpose | Example |
|---|---|---|
| `upper()` | Converts the string to uppercase | `"python".upper()` |
| `lower()` | Converts the string to lowercase | `"PYTHON".lower()` |
| `capitalize()` | Converts the first character to uppercase | `"python".capitalize()` |
| `title()` | Converts the first character of each word to uppercase | `"hello world".title()` |
| `strip()` | Removes spaces from both ends | `" Hello ".strip()` |
| `replace()` | Replaces a part of the string | `"Hello".replace("H", "J")` |
| `split()` | Splits the string into a list | `"Hello World".split()` |
| `find()` | Returns the position of a substring | `"Python".find("t")` |
| `count()` | Returns the number of occurrences | `"Python".count("o")` |

### len, max, min, sorted

````python
c = "India"

print(len(c))
print(max(c)) # letter with maximun ascii value
print(min(c)) # letter with minimum ascii value
print(sorted(c)) # sorted on the basis of ascii value
print(sorted(c, reverse=True))  # decending order
````

### capitalize, title, upper, lower, swapcase

````python
a = "heLlo woRld   "

print(a.capitalize()) # first character to uppercase
print(a.title())      # first character of each word to uppercase
print(a.upper())
print(a.lower())
print(a.swapcase())  # uppercase to lowercase and vice-versa
````
