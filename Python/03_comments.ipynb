# Comments

Comments are used to **add explanations or notes** to the code. Python ignores comments while executing the program.

### Single-Line Comment

A single-line comment starts with `#`.

```python
# This is a comment
print("Hello World")


### Multi-Line Comments

Python does not have a dedicated syntax for multi-line comments.

To write a multi-line comment, use `#` at the beginning of each line.

```python
# This is a multi-line comment
# Each line starts with the
# hash symbol
print("Hello World")

### Triple-Quoted Strings

Triple quotes (`'''` or `"""`) can be used to write text across multiple lines.

```python
'''
This is a multi-line string.
It can span multiple lines.
'''

"""
This is also a multi-line string.
"""
