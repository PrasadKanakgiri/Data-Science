# Data Types

Data types specify the **type of data** that a variable can store.

Python data types can be broadly classified into **3 types**:

## 1. Basic Types

Basic data types are used to store a single value.

- `int` → Integer
- `float` → Floating-point number
- `complex` → Complex number
- `bool` → Boolean
- `str` → String

## 2. Container Types

Container data types are used to store **multiple values**.

- `list` → List
- `tuple` → Tuple
- `set` → Set
- `dict` → Dictionary

## 3. User-Defined Types

User-defined data types are created by the programmer.

- `class` → Class

# Basic Types of Data Types

Python has different types of data. The basic data types include:

- `int` → Integer
  - Represents whole numbers without a decimal point.
  - Python integers can be **arbitrarily large** and do not have a fixed maximum value (limited by available memory).

- `float` → Floating-point number
  - Represents numbers with a decimal point.
  - The largest finite `float` is approximately **1.8 × 10³⁰⁸**.
  - Numbers larger than this are represented as `inf` (infinity).

- `bool` → Boolean
  - Represents either `True` or `False`.

- `complex` → Complex number
  - Represents numbers with a real and imaginary part.
  - Example: `5 + 4j`

- `str` → String
  - Strings can be written using single quotes (`' '`), double quotes (`" "`), or triple quotes (`''' '''`).

```python
print(5)

print(1e308)
print(1e309)

print(3.5)
print(1.7e308)
print(1.7e309)

print(True)
print(False)

print(5+4j)

print('India')
print("India")
print('''India''')

````python
print(3.5)
print(1.7e308)
print(1.7e309)
````

````python
print(True)
print(False)
````

````python
print(5+4j)
````

````python
print('India')
print("India")
print('''India''')
````

## 2. Container Types of Data Types

Container data types are used to store **multiple values** in a single object.

### List

A `list` is an **ordered and mutable** collection of values.


### Tuple

A `tuple` is an **ordered and immutable** collection of values.


### Dictionary

A `dictionary` is a collection of **key-value pairs**.


### Set

A `set` is an **unordered collection of unique values**.

```python
print([1, 2, 3, 4, 5])                               # List
print((1, 2, 3, 4, 5))                               # Tuple
print({1, 2, 3, 4, 5})                               # Sets
print({"Name": "John", "Age": 10, "Gender": "Male"}) # Dictionary



````python
print([1,2,3,4,5])
print((1,2,3,4,5))
print({1,2,3,4,5})
print({"Name": "John", "Age": 10, "Gender": "Male"})
````
