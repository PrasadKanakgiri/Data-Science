# Multiplication

```python
# 1st approach
def helper(a, b, result):
    if b == 0:
        return result
    
    result = result + a
    return helper(a, b-1, result)
    
def multiply(a, b):
    result = 0
    ans = helper(a, b, result)
    return ans

result = multiply(9, 4)
print(result)
```

```python
# 2nd approach
def multiply(a, b):
    if b == 0:
        return 0
    
    return a + multiply(a, b-1)

result = multiply(9, 4)
print(result)
```

# Factorial

```python
def factorial(a):
    if a == 1:
        return 1
    
    return a * factorial(a-1)

result = factorial(5)
print(result)
```

# Palindrome

```python
# 1 approach -> with recursion
def palindrome(s):
    if len(s) <= 1:
        return "Palindrome"
    else:
        if s[0] == s[-1]:
            return palindrome(s[1:-1])
        else:
            return "Not a Palindrome"
        
print(palindrome("abba"))
```

```python
# 2 approach -> without recursion

s = "121212121"

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
def func(months, rabbits):
    if months == 12:
        return rabbits
    
    rabbits = rabbits * 2
    
    return rabbits + func(months+1, rabbits)

def main(months):
    rabbits = 0
    
    ans = func(months, rabbits)
    return ans
print(main(1))
```

# Fibonacci Series

```python
def fibonacci(num):
    if num == 0 or num == 1:
        return 1
    
    return fibonacci(num-1) + fibonacci(num-2)

print(fibonacci(8))
```

# Power Set

```python
def helper(nums, index, current, result):
    if index == len(nums):
        result.append(current.copy())
        return
    
    # Dont include nums[index]
    helper(nums, index+1, current, result)
    
    # Include nums[index]
    current.append(nums[index])
    helper(nums, index+1, current, result)
    
    # Backtrack
    current.pop()
    
    
def power_set(nums):
    result = []
    index = 0
    current = []
    helper(nums, index, current, result)
    
    return result

nums = [1, 2, 3]
print(power_set(nums))
```
