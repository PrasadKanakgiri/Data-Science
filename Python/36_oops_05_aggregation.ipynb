# Aggregation in Python

**Aggregation** is an OOP relationship where **one class uses another class, but both objects can exist independently**.

In simple words:

> **Aggregation = HAS-A relationship, where the contained object can exist independently.**

For example:

```text
ATM has a Bank
```

But the **Bank can exist without the ATM**, and the **ATM can exist without that particular Bank object**.

---

# ATM Example

Let's create a `Bank` class:

```python
class Bank:

    def __init__(self, name):
        self.name = name

    def show_bank(self):
        print("Bank:", self.name)
```

Now create an `Atm` class:

```python
class Atm:

    def __init__(self, bank):
        self.bank = bank

    def show_atm_details(self):
        print("ATM is connected to:", self.bank.name)
```

Now create the objects:

```python
bank1 = Bank("SBI")

atm1 = Atm(bank1)

atm1.show_atm_details()
```

Output:

```text
ATM is connected to: SBI
```

---

# What is happening here?

First we create a `Bank` object:

```python
bank1 = Bank("SBI")
```

Then we pass that existing object to `Atm`:

```python
atm1 = Atm(bank1)
```

Inside `Atm`:

```python
self.bank = bank
```

So `atm1` stores a **reference to the existing ****`bank1`**** object**.

Conceptually:

```text
bank1
  ↓
┌──────────────┐
│ Bank         │
│ name = SBI   │
└──────────────┘
       ↑
       │
   self.bank
       │
┌──────────────┐
│ atm1         │
│              │
│ bank ────────┘
└──────────────┘
```

The ATM **has a reference to a Bank object**.

That's the **HAS-A relationship**.

---

# Why is this Aggregation?

Because `bank1` can exist independently:

```python
bank1 = Bank("SBI")
```

Even if we don't create an ATM, the bank still exists.

```python
bank1.show_bank()
```

Output:

```text
Bank: SBI
```

And the ATM can also be created using another bank:

```python
bank2 = Bank("HDFC")

atm2 = Atm(bank2)
```

So:

```text
Bank ← independent
ATM  ← independent

ATM HAS-A Bank
```

---

# Important Difference: Aggregation vs Instance Variable

You might think:

```python
self.bank = bank
```

is just an instance variable.

It **is** an instance variable, but it also represents an **aggregation relationship** between two classes.

For example:

```python
class Atm:

    def __init__(self, bank):
        self.bank = bank
```

Here:

```text
self.bank
```

is an **instance variable**.

But because it stores an object of another class:

```text
Atm → Bank
```

it represents an **aggregation relationship**.

---

# Real-World Example

Think about:

```text
Department HAS-A Teacher
```

A teacher can exist without the department.

```python
class Teacher:

    def __init__(self, name):
        self.name = name


class Department:

    def __init__(self, teacher):
        self.teacher = teacher


teacher1 = Teacher("Rahul")

department1 = Department(teacher1)
```

Here:

```text
Teacher
   ↑
   │
Department
```

The department uses the teacher, but the teacher can exist independently.

Therefore, this is **Aggregation**.

---

### One-line definition

> **Aggregation is a HAS-A relationship where one class contains a reference to an object of another class, while both objects can exist independently.**


````python
# Example 1

class Bank:
    
    def __init__(self, name):
        self.name = name
    
    def show_bank(self):
        print("Bank: ", self.name)

class Atm:

        def __init__(self, bank):
            self.bank = bank
        
        def show_bank_details(self):
            print("Atm is connected to:", self.bank.name)

bank1 = Bank("SBI")
atm1 = Atm(bank1)

atm1.show_bank_details()
````

````python
# Example 2

class Department:
    
    def __init__(self, dept_name):
        self.dept_name = dept_name
        
    def show_teacher_name(self):
        print("Teacher Name:", self.dept_name.teacher_name)

class Teacher:
    
    def __init__(self, teacher_name):
        self.teacher_name = teacher_name
        
teacher1 = Teacher("Ankita")
department1 = Department(teacher1)
department1.show_teacher_name()
````

````python
# Example 3

class Address:
    
    def __init__(self, city, pincode, state):
        self.city = city
        self.pincode = pincode
        self.state = state
        
    def change_address(self, new_city, new_pincode, new_state):
        self.city = new_city
        self.pincode = new_pincode
        self.state = new_state
        
class Customer:
    
    def __init__(self, name, gender, address):
        self.name = name
        self.gender = gender
        self.address = address
        
    def show_customer_details(self):
        print("Customer city is:", self.address.city)
        print("Pincode is:", self.address.pincode)
        print("State is:", self.address.state)
    
    def edit_profile(self, new_name, new_city, new_pincode, new_state):
        self.name = new_name
        self.address.change_address(new_city, new_pincode, new_state)
    
    

address1 = Address("Sangli", 416416, "Maharashtra")
cust1 = Customer("Nitish", "Male", address1)

cust1.edit_profile("Ankita", "Bangalore", "511010", "Karnataka")

cust1.show_customer_details()
````
