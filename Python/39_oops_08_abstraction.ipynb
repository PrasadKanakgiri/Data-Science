# Abstraction

**Abstraction** means:

> **Showing only the important details and hiding the unnecessary implementation details.**

In simple words:

> **Tell the user WHAT to do, but hide HOW it is done.**

---

## Real-Life Example

When you use an ATM:

```text
You:
  → Enter PIN
  → Select "Withdraw"
  → Enter amount
```

You don't need to know:

* How the ATM communicates with the bank
* How the account balance is checked
* How the transaction is processed
* How the cash dispenser works

You only see the **necessary operations**.

That's abstraction.

---

# Abstraction in Python

Python provides abstraction using **abstract classes and abstract methods**.

We use the `abc` module:

```python
from abc import ABC, abstractmethod
```

---

## Simple Example

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
```

Here:

```python
@abstractmethod
def start(self):
    pass
```

is an **abstract method**.

It says:

> Every child class must provide its own implementation of `start()`.

---

## Child Classes

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts with a key")


class Bike(Vehicle):

    def start(self):
        print("Bike starts with a button")
```

Now:

```python
car = Car()
bike = Bike()

car.start()
bike.start()
```

Output:

```text
Car starts with a key
Bike starts with a button
```

The parent class says:

```text
Vehicle
   ↓
Every vehicle MUST have start()
```

But it doesn't specify exactly how it starts.

The child classes decide that.

---

# Why Use Abstraction?

Suppose we have:

```text
Vehicle
   |
   ├── Car
   ├── Bike
   └── Truck
```

We know that every vehicle should have:

```python
start()
```

But the implementation can be different:

```text
Car   → start with key
Bike  → start with button
Truck → start with key/card
```

So the parent class defines **WHAT is required**:

```python
start()
```

and the child class defines **HOW it works**.

---

# Abstract Class

A class containing at least one abstract method is called an **abstract class**.

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
```

`Vehicle` is an abstract class.

You **cannot normally create an object directly** from it:

```python
vehicle = Vehicle()
```

This gives an error because `start()` has not been implemented.

---

# Abstract Method

An abstract method is a method that is declared in the parent class but **must be implemented by the child class**.

```python
@abstractmethod
def start(self):
    pass
```

The parent basically says:

> "I'm defining that this method must exist, but the child decides how it works."

---

# Abstraction vs Encapsulation

These two are often confused.

### Abstraction

**Hides implementation details.**

```text
WHAT should happen?
        ↓
start()
```

### Encapsulation

**Controls access to data.**

```text
How should data be accessed?
        ↓
__balance
```

For example:

```text
Abstraction  → Hide HOW something works
Encapsulation → Control access to data
```

### Easy memory trick

> **Abstraction = WHAT**
> **Implementation = HOW**

Example:

```text
Vehicle says:
"Every vehicle must have start()."
             ↓
       WHAT

Car says:
"Here is how I start."
             ↓
       HOW
```


````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    
    def start(self):
        print("Car starts with a key")
        
class Bike(Vehicle):
    
    def start(self):
        print("Bike starts with a button")
        
vehicle = Vehicle()
vehicle.start()
````
