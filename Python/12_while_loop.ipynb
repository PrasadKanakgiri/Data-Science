# while Loop

A `while` loop is used to **repeat a block of code as long as a condition is `True`**.

### Syntax

```python
while condition:
    # code to execute

```
### Example
```python
i = 1

while i <= 5:
    print(i)
    i = i + 1
```
### Output
1
2
3
4
5

# Looping in Real-World Applications

Loops are commonly used in applications to **process or display multiple items**.

For example, when we search for mobile phones on an e-commerce app like Flipkart, many mobile products need to be displayed.

Instead of writing code separately for every mobile, a loop can be used to **process each mobile one by one**.

### Example

```python
mobiles = ["iPhone 15", "Samsung S24", "OnePlus 12", "Pixel 9"]

for mobile in mobiles:
    print(mobile)
```

### Output:

iPhone 15
Samsung S24
OnePlus 12
Pixel 9

The loop goes through each mobile one by one.

### How It Works in an E-Commerce App

Conceptually, the application might work like this:

```python
mobiles = get_mobiles_from_database()

for mobile in mobiles:
    display_mobile(mobile)
```

The loop:

Gets the first mobile → displays it.

Gets the second mobile → displays it.

Gets the third mobile → displays it.

Continues until all required mobiles are displayed.

Loops help avoid writing the same code repeatedly for every product.

Note: Flipkart may use different programming languages and technologies. This is a simplified Python example to understand how loops are used in real-world applications

````python
num = int(input("Enter your number: "))

i = 1

while i <= 10:
    print(num, "x", i, "=", num * i)
````
