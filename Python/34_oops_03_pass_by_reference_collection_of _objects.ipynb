# Collection of Objects

```python
class Customer:
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def intro(self):
        print("Hi I am ", self.name, "I am ", self.age)

        
cust1 = Customer("Nitish", 24)
cust2 = Customer("Ankita", 27)
cust3 = Customer("Rahuk", 28)

L = [cust1, cust2, cust3]

for i in L:
    print(i.name, i.age)
    i.intro()

```
