# Class in Python

## Definition

A `class` is a blueprint or template used to create **objects**.

A class defines:
- **Attributes** → data/properties of an object
- **Methods** → functions/behaviors that an object can perform

Think of a class as a **design/template**, and an object as the **actual thing created from that template**.

### Real-World Example

Suppose we want to represent students.

A `Student` class can define:
- `name`
- `age`
- `roll_no`

And methods such as:
- `study()`
- `attend_class()`

The class is the blueprint, while individual students are objects created from it.

## Syntax

```python
class ClassName:
    # attributes
    # methods
    pass
```
The `class` keyword is used to create a class.

### Example
```python
class Student:
    name = "Prasad"
    age = 25

    def study(self):
        print("Student is studying")
```
##### Here:

Student → class

name and age → attributes

study() → method

````python
class Student:
    name = "Prasad"
    age = 23
    
    def study(self):
        print("Student is Studying")
````

# Creating an Object in Python

## Definition

An **object** is an instance of a class.

After defining a class, we can create objects from that class. This process is called **object creation** or **instantiation**.

## Syntax

```python
object_name = ClassName()
```
- `object_name` → name of the object
- `ClassName` → name of the class
- `()` → calls the class to create an object

### Example
```python
class Student:
    name = "Prasad"
    age = 25

student1 = Student()
```
Here:

- `Student` is the class.
- `student1` is the object.
- `Student()` creates an object of the Student class.

## Accessing Attributes

We can access the attributes of an object using the dot `(.)` operator.

```python
print(student1.name)
print(student1.age)
```
### Output
Prasad
25

## Creating Multiple Objects

We can create multiple objects from the same class.
```python
student1 = Student()
student2 = Student()
student3 = Student()
```
**All three objects are instances of the `Student` class.**

### Key Points
- An object is an instance of a class.
- Objects are created using ClassName().
- The process of creating an object is called instantiation.
- The . operator is used to access an object's attributes and methods.
- One class can have multiple objects.

````python
class Student():
    name = "Prasad"
    age = 23
    
    def study(self):
        return "Student is Studying"

s1 = Student()
s2 = Student()
s3 = Student()

age = s1.age
name = s2.name
str1 = s3.study()    # s3.study() will execute the method
str2 = s3.study      # s3.study will store the method

print(age)
print(str1)
print(str2())
print(name)
````

# Instance Variable in Python

## Definition

An **instance variable** is a variable that belongs to a **specific object (instance)** of a class.

Each object has its **own copy** of the instance variables.

Instance variables are usually created using `self`.

```python
self.variable_name = value
```
### Example
```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```
Here:
```python
self.name
self.age
```
are instance variables.

They belong to the individual `objects` created from the `Student` `class`.

### Creating Objects
```python
student1 = Student("Prasad", 25)
student2 = Student("Rahul", 22)
```
Now each `object` has its own instance variables:



Even though both `objects` were created from the same `class`, their **instance variables** contain different values.

# Constructor in Python

## Definition

A **constructor** is a special method that is automatically called when an object is created.

In Python, the constructor is usually the `__init__()` method.

It is mainly used to **initialize the attributes of an object** with values.

---

## Syntax

```python
class ClassName:

    def __init__(self):
        # initialization code
        pass
```
### When we create an object:
```python
object1 = ClassName()
```
Python automatically calls:
```python
__init__()
```

### Example
```python
class Student:

    def __init__(self):
        self.name = "Prasad"
        self.age = 25

student1 = Student()

print(student1.name)
print(student1.age)
```
### Output

Prasad
25

### What happens?

When this statement executes:
```python
student1 = Student()
```
**Python creates the object and automatically calls:**
```python
__init__()
```
So these statements are executed:
```python
self.name = "Prasad"
self.age = 25
```
The object's attributes are therefore initialized with those values.

## Constructor with Parameters

We can pass values to the constructor while creating the object.
```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Prasad", 25)
student2 = Student("Rahul", 22)

print(student1.name)
print(student1.age)

print(student2.name)
print(student2.age)
```
### Output
Prasad
25
Rahul
22

Here:
```python
student1 = Student("Prasad", 25)
```
automatically calls:
```python
__init__("Prasad", 25)
```

## Why Do We Need a Constructor?

Without a constructor, we would have to manually assign values after creating the object:
```python
class Student:
    pass

student1 = Student()

student1.name = "Prasad"
student1.age = 25
```
With a constructor, we can initialize the object directly:
```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Prasad", 25)
```
This makes object creation cleaner and more organized.

````python
class Student():
    
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Prasad", 23)
s2 = Student("Nitish", 25)

print(s1.name)
print(s2.age)
````

# `self` in Python

## What is `self`?

`self` refers to the **current object**.

In simple words:

> **`self` tells Python which object we are talking about.**

This is especially important when we create multiple objects from the same class.

---

## Simple Example

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```
### Now let's create two objects:
```python
student1 = Student("Prasad", 25)
student2 = Student("Rahul", 22)
```
Here, we have two different objects:

student1 → name = "Prasad", age = 25

student2 → name = "Rahul", age = 22

But both objects use the same `class` and the same `__init__()` method.

So how does Python know where to store `"Prasad"` and `"Rahul"`?

That's where `self` comes in.

# How `self` Works

When we write:
```python
student1 = Student("Prasad", 25)
```
Python internally passes `student1` to `__init__()`.

Conceptually, it is like:
```python
Student.__init__(student1, "Prasad", 25)
```
Therefore, inside the constructor:

`self`

refers to:

`student1`

So:
```python
self.name = name
```
becomes:
```python
student1.name = "Prasad"
```
And:
```python
self.age = age
```
becomes:
```python
student1.age = 25
```

### The Most Important Point

**The value of self changes depending on which object is calling the method.**

# Why Do We Write  `self.name` Instead of Just `name`?

This is very important.

Consider:
```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```
There are actually two different things called `name` here.

1. `name`
```python
name
```
This is a parameter.

**It temporarily contains the value passed to the constructor.**

For example:
```python
Student("Prasad", 25)
```
means:

name → "Prasad"
age  → 25

2. `self.name`
```python
self.name
```
This is an attribute belonging to the current object.

So:

self.name = name

means:

Take the value from the parameter `name`
                
                ↓
        
            "Prasad"
                
                ↓

Store it in the current object's `name` attribute

Therefore:

`name`       → temporary parameter

`self.name`  → object's attribute

````python
class Atm:
    
    def __init__(self):
        self.pin = ""
        self.balance = 0
        
        self.menu()
    
    def menu(self):
        user_input = input(
            '''
            1. Enter '1' to create pin.
            2. Enter '2' to deposit
            3. Enter '3' to withdraw
            4. Enter '4' to check balance
            5. Enter '5' to exit
            '''
        )
        
        if user_input == '1':
            self.create_pin()
        elif user_input == '2':
            self.deposit()
        elif user_input == '3':
            self.withdraw()
        elif user_input == '4':
            self.check_balance()
        elif user_input == '5':
            print("Bye")
        else:
            print("Invalid option")
    
    def create_pin(self):
        self.pin = input("Enter your pin: ")
        print("Pin created successfully")
    
    def deposit(self):
        enter_pin = input("Enter your pin: ")
        
        if enter_pin == self.pin:
            amount = int(input("Enter amount to deposit: "))
            self.balance = self.balance + amount
            print("Deposit Successful")
        else:
            print("Incorrect pin")
    
    def withdraw(self):
        enter_pin = input("Enter your pin: ")
        
        if enter_pin == self.pin:
            amount = int(input("Enter amount to withdraw: "))
            
            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Withdraw Successful")
            else:
                print("Insufficient Funds")
        else:
            print("Incorrect pin")
    
    def check_balance(self):
        enter_pin = input("Enter your pin: ")
        
        if enter_pin == self.pin:
            print("Balance:", self.balance)
        else:
            print("Incorrect pin")


sbi = Atm()
````

````python
sbi.deposit()
print(sbi.balance)
````

````python
hdfc = Atm()
````

````python
hdfc.deposit()
hdfc.withdraw()
hdfc.check_balance()
````

# `__add__()` in Python

## Definition

`__add__()` is a **special method** (also called a **magic method** or **dunder method**) in Python.

It defines what should happen when the `+` operator is used with objects.

For example:

```python
a + b
`__add__` is another magic method which executed when there is `+` operation in a print statement
```
# Why Do We Use `__add__()` in OOP?

Suppose we create our own class:

```python
class Number:

    def __init__(self, value):
        self.value = value
```
Now create two objects:
```python
a = Number(10)
b = Number(20)
```
If we try:
```python
print(a + b)
```
Python doesn't automatically know how two Number objects should be added.

We can define that behavior using `__add__()`.

### Defining `__add__()`
```python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
```
Now:
```python
a = Number(10)
b = Number(20)

print(a + b)
```
### Output

30

````python
class Fraction:
    
    def __init__(self, n, d):
        self.num = n
        self.den = d
    
    def __str__(self):
        return "{}/{}".format(self.num, self.den)
    
    def __add__(self, other):  # self is 1st fraction and other is 2nd fraction
        temp_num = self.num * other.den + other.num * self.den
        temp_den = self.den * other.den
        
        return "{}/{}".format(temp_num, temp_den)
    
    def __sub__(self, other):
        temp_num = self.num * other.den - other.num * self.den
        temp_den = self.den * other.den
        
        return "{}/{}".format(temp_num, temp_den)
    
    def __mul__(self, other):
        temp_num = self.num * other.num
        temp_den = self.den * other.den
        
        return "{}/{}".format(temp_num, temp_den)
    
    def __truediv__(self, other):
        temp_num = self.num * other.den
        temp_den = self.den * other.num
        
        return "{}/{}".format(temp_num, temp_den)
        

x = Fraction(5, 6)
print(x)    # __str__ is called after print(x) is executed

y = Fraction(6, 5)
print(y)

z = x + y   # __add__ is called here as there is a '+' operation
print(z)

p = x - y   # __sub__ is called here as there is a '-' operation
print(p)

q = x * y   # __mul__ is called here as there is a '*' operation
print(q)

r = x / y   # __truediv__ is called here as there is a '/' operation
print(z)
````
