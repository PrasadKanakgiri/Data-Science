# Dictionary

A `dictionary` is a **mutable collection of key-value pairs**.

Dictionaries are created using **curly braces `{}`**.

Each item contains a **key** and its corresponding **value**.

#### Important Points
- Dictionary is mutable.
- Dictionary stores data as key-value pairs.
- Keys must be unique.
- Keys must be hashable.
- Values can be of different data types.
- Dictionaries are ordered in modern Python.

```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

print(student)
```
### Output:

{'name': 'John', 'age': 25, 'city': 'Pune'}

### Key-Value Pair

A dictionary stores data in the form:

key : value

### For example:
```python
student = {
    "name": "John",
    "age": 25
}
```
Here:

"name" → key

"John" → value

"age" → key

25 → value

### Creating a Dictionary
```python
person = {
    "name": "Alice",
    "age": 22,
    "is_student": True
}

print(person)
```
A dictionary can contain different data types as values.

We can also use `dict()`.
```python
student = dict()

print(student)

````python
person1 = {
    "name": "Alice",
    "age" : 22,
    "is_student" : True
}

print(person1)

person2 = dict(
    name = "Bob",
    age = 23,
    is_student = False
)

print(person2)
````

### Accessing Values

We can access a `value` using its `key`.
```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

print(student["name"])
print(student["age"])
```
### Output:

John
25

````python
person1 = {
    "name": "Alice",
    "age" : 22,
    "is_student" : True
}

print(person1["name"])
````

### Editing Values

Dictionaries are **mutable**, so we can change the value of an existing key.

```python
student = {
    "name": "John",
    "age": 25
}

student["age"] = 30

print(student)
```
### Output:

{'name': 'John', 'age': 30}

````python
student = {
    "name": "John",
    "age": 25
}

student["name"] = "Alex"

print(student)
````

### get()

The `get()` method is used to **access the value of a key** in a dictionary.

```python
student = {
    "name": "John",
    "age": 25
}

print(student.get("name"))
print(student.get("age"))
```
### Output:

John
25

If the key does not exist, `get()` returns `None` instead of giving an error.
```python
student = {
    "name": "John",
    "age": 25
}

print(student.get("city"))
```
### Output:

None

We can also provide a `default` value if the key does not exist.
```python
print(student.get("city", "Pune"))
```
### Output:

Pune

### Difference Between [] and get()
```python
student = {
    "name": "John",
    "age": 25
}

print(student["city"])
```

This gives a `KeyError` because "city" does not exist.
```python
print(student.get("city"))
```
This returns:

None

So, `get()` is useful when we want to access a key safely without getting an error if the key is missing.

````python
student = {
    "name": "John",
    "age": 25
}

print(student.get("name"))

print(student.get("city"))
print(student.get("city", "Pune"))

# print(student["city"])   # this will throw error as "city" is not present in the dictionary
````

### Adding New Key-Value Pairs

A new key-value pair can be added to a dictionary by **assigning a value to a new key**.

```python
student = {
    "name": "John",
    "age": 25
}

student["city"] = "Pune"

print(student)
```
### Output:

{'name': 'John', 'age': 25, 'city': 'Pune'}

Here, `"city"` is the new key and `"Pune"` is its value.

### update()
**We can also add multiple key-value pairs using `update()`.**
```python
student = {
    "name": "John",
    "age": 25
}

student.update({
    "city": "Pune",
    "gender": "Male"
})

print(student)
```
### Output:

{'name': 'John', 'age': 25, 'city': 'Pune', 'gender': 'Male'}

**If the key already exists, assigning a new value will update the existing value instead of creating a duplicate key.**

````python
student = {
    "name": "John",
    "age": 25
}

print(student)

student["city"] = "Pune"
student["gender"] = "Male"

print(student)

student.update({
    "roll-no" : 21,
    "is_present" : True
})

print(student)
````

### Deleting

There are different ways to delete elements from a dictionary.

### Using del

The `del` keyword is used to **delete a specific key-value pair**.

```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

del student["age"]

print(student)
```
### Output:

{'name': 'John', 'city': 'Pune'}

### pop()

The `pop()` method is used to **remove a key-value pair and return its value**.
```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

age = student.pop("age")

print(age)
print(student)
```
### Output:

25
{'name': 'John', 'city': 'Pune'}

### popitem()

The `popitem()` method **removes and returns the last inserted key-value pair**.
```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

item = student.popitem()

print(item)
print(student)
```
### Output:

('city', 'Pune')
{'name': 'John', 'age': 25}

### clear()

The `clear()` method is used to **remove all key-value pairs from the dictionary**.
```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

student.clear()

print(student)
```
### Output:

{}

````python
student = {
    "name": "John",
    "age": 25,
    "city" : "Pune"
}

print(student)

del student["city"]
print(student)

student = {
    "name": "John",
    "age": 25,
    "city" : "Pune"
}

student.pop("age")
print(student)

student.popitem()
print(student)

student.clear()
print(student)
````

### Dictionary Operations

Dictionary operations are used to **access, add, update, remove, and check key-value pairs**.

### Accessing Values

We can access a value using its key.

```python
student = {
    "name": "John",
    "age": 25
}

print(student["name"])
```
### Output:

John

### Membership

The in operator checks whether a key exists in the dictionary.
```python
student = {
    "name": "John",
    "age": 25
}

print("name" in student)
print("city" in student)
```
### Output:

True
False

### Length

The len() function returns the number of key-value pairs.
```python
student = {
    "name": "John",
    "age": 25,
    "city": "Pune"
}

print(len(student))
```
### Output:

3

### Getting Keys

The keys() method returns all the keys in the dictionary.
```python
student = {
    "name": "John",
    "age": 25
}

print(student.keys())
```
### Getting Values

The values() method returns all the values in the dictionary.
```python
student = {
    "name": "John",
    "age": 25
}

print(student.values())
```
Getting Key-Value Pairs

The items() method returns all the key-value pairs.
```python
student = {
    "name": "John",
    "age": 25
}

print(student.items())
```
### Concatenation

Dictionaries cannot be concatenated using the + operator.
```python
dict1 = {"a": 1}
dict2 = {"b": 2}
```
 print(dict1 + dict2)  # TypeError

To combine dictionaries, we can use update().
```python
dict1 = {"a": 1}
dict2 = {"b": 2}

dict1.update(dict2)

print(dict1)
```
### Output:

{'a': 1, 'b': 2}

````python
student = {
    "name": "John",
    "age": 25
}

print(student["name"])

print("name" in student)
print("city" in student)

print(len(student))

print(student.keys())
print(student.values())
print(student.items())

dict1 = {"a": 1}
dict2 = {"b": 2}

# print(dict1 + dict2)  # TypeError

dict1.update(dict2)
print(dict1)
````
